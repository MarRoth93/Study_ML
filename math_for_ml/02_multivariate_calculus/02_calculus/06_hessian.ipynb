# The Hessian Matrix

## Introduction

Now that we've explored the Jacobian, which collects all **first-order partial derivatives** of a function into a vector, we can extend this concept to **second-order derivatives**.

The **Hessian matrix** collects all second-order partial derivatives of a multivariable function into a matrix. This powerful tool helps us understand the curvature of functions and identify the nature of critical points (maxima, minima, or saddle points).

For a function $f(x_1, x_2, ..., x_n)$ of $n$ variables, the Hessian is an $n \times n$ square matrix:

$$
H = \begin{bmatrix}
\frac{\partial^2 f}{\partial x_1^2} & \frac{\partial^2 f}{\partial x_1 \partial x_2} & \cdots & \frac{\partial^2 f}{\partial x_1 \partial x_n} \\
\frac{\partial^2 f}{\partial x_2 \partial x_1} & \frac{\partial^2 f}{\partial x_2^2} & \cdots & \frac{\partial^2 f}{\partial x_2 \partial x_n} \\
\vdots & \vdots & \ddots & \vdots \\
\frac{\partial^2 f}{\partial x_n \partial x_1} & \frac{\partial^2 f}{\partial x_n \partial x_2} & \cdots & \frac{\partial^2 f}{\partial x_n^2}
\end{bmatrix}
$$

## Understanding Second-Order Partial Derivatives

Just as we can keep differentiating a single-variable function to find higher-order derivatives, we can do the same with partial derivatives.

**Second-order partial derivative** with respect to $x_1$ then $x_2$:
1. First, differentiate $f$ with respect to $x_1$ (treating all other variables as constants)
2. Then, differentiate the result with respect to $x_2$ (again treating all other variables as constants)

This is written as: $\frac{\partial^2 f}{\partial x_2 \partial x_1}$ or $\frac{\partial}{\partial x_2}\left(\frac{\partial f}{\partial x_1}\right)$

**Key Property:** For continuous functions, the order of differentiation doesn't matter:
$$\frac{\partial^2 f}{\partial x_i \partial x_j} = \frac{\partial^2 f}{\partial x_j \partial x_i}$$

This means the Hessian matrix is **symmetric** across its diagonal!

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import sympy as sp
from matplotlib import cm

# For better visualization
plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline
```

## Example 1: Computing the Hessian for $f(x, y, z) = x^2yz$

Let's work through a concrete example step by step. 

**Strategy:** It's often easier to:
1. First find the **Jacobian** (vector of first derivatives)
2. Then differentiate each term again to build the **Hessian** (matrix of second derivatives)

### Step 1: Find the Jacobian

For $f(x, y, z) = x^2yz$, we compute all first-order partial derivatives:

```python
# Define symbolic variables
x, y, z = sp.symbols('x y z')

# Define the function
f = x**2 * y * z

print("Function f(x, y, z) =", f)
print("\n" + "="*50)
print("JACOBIAN (First-order partial derivatives):")
print("="*50)

# Compute the Jacobian (first derivatives)
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)
df_dz = sp.diff(f, z)

print(f"∂f/∂x = {df_dx}")
print(f"∂f/∂y = {df_dy}")
print(f"∂f/∂z = {df_dz}")

# Jacobian as a vector
J = sp.Matrix([df_dx, df_dy, df_dz])
print(f"\nJacobian J = {J.T}")
```

### Step 2: Build the Hessian Matrix

Now we differentiate each term in the Jacobian with respect to each variable:

- **Row 1:** Differentiate $\frac{\partial f}{\partial x} = 2xyz$ with respect to $x$, $y$, and $z$
- **Row 2:** Differentiate $\frac{\partial f}{\partial y} = x^2z$ with respect to $x$, $y$, and $z$
- **Row 3:** Differentiate $\frac{\partial f}{\partial z} = x^2y$ with respect to $x$, $y$, and $z$

```python
# Compute the Hessian matrix (second derivatives)
print("="*50)
print("HESSIAN (Second-order partial derivatives):")
print("="*50)

# Build Hessian manually to show the process
H = sp.Matrix([
    [sp.diff(df_dx, x), sp.diff(df_dx, y), sp.diff(df_dx, z)],
    [sp.diff(df_dy, x), sp.diff(df_dy, y), sp.diff(df_dy, z)],
    [sp.diff(df_dz, x), sp.diff(df_dz, y), sp.diff(df_dz, z)]
])

print("\nHessian matrix H =")
print(H)

print("\n" + "="*50)
print("Notice the SYMMETRY:")
print("="*50)
print(f"H[0,1] = {H[0,1]} = H[1,0] = {H[1,0]}")
print(f"H[0,2] = {H[0,2]} = H[2,0] = {H[2,0]}")
print(f"H[1,2] = {H[1,2]} = H[2,1] = {H[2,1]}")
print("\nThe Hessian is symmetric! This is always true for continuous functions.")
```

### Evaluating the Hessian at a Point

Now we can pass specific coordinates $(x, y, z)$ into our Hessian, and it will return a matrix of numbers that tells us about the curvature at that point.

```python
# Evaluate at a specific point, e.g., (1, 2, 3)
point = {x: 1, y: 2, z: 3}
H_at_point = H.subs(point)

print(f"Hessian evaluated at point (x=1, y=2, z=3):")
print(H_at_point)
```

## Using the Hessian to Identify Critical Points

The real power of the Hessian comes when we want to classify critical points (where the gradient is zero).

### The Problem

Suppose you find a point where the **Jacobian (gradient) is zero**. How do you know if it's:
- A **minimum**? 
- A **maximum**?
- A **saddle point**?

You could check nearby points, but that's not robust. Instead, we use the **Hessian**!

### The Hessian Test

At a critical point (where gradient = 0), we examine the Hessian:

1. **Calculate the determinant** of the Hessian matrix
2. **Check the first element** (top-left corner)

The rules are:
- If $\det(H) > 0$ and $H_{11} > 0$ → **Local minimum**
- If $\det(H) > 0$ and $H_{11} < 0$ → **Local maximum**  
- If $\det(H) < 0$ → **Saddle point**
- If $\det(H) = 0$ → Test is inconclusive

Let's see this in action with 2D examples (easier to visualize)!

## Example 2: Local Minimum - $f(x, y) = x^2 + y^2$

This is a simple paraboloid opening upward (like a bowl). We expect a minimum at $(0, 0)$.

```python
# Define new variables for 2D case
x, y = sp.symbols('x y')

# Define function: f(x,y) = x^2 + y^2
f_min = x**2 + y**2

print("Function: f(x, y) =", f_min)
print("\n" + "="*50)
print("STEP 1: Find the Jacobian (gradient)")
print("="*50)

# Jacobian
J_min = sp.Matrix([sp.diff(f_min, x), sp.diff(f_min, y)])
print("Jacobian J =", J_min.T)

# Evaluate at (0, 0)
J_at_origin = J_min.subs({x: 0, y: 0})
print(f"\nAt point (0, 0): J = {J_at_origin.T}")
print("→ Gradient is ZERO! This is a critical point.")

print("\n" + "="*50)
print("STEP 2: Compute the Hessian")
print("="*50)

# Hessian
H_min = sp.hessian(f_min, [x, y])
print("Hessian H =")
print(H_min)

print("\n" + "="*50)
print("STEP 3: Apply the Hessian Test")
print("="*50)

# Evaluate at (0, 0)
H_at_origin = H_min.subs({x: 0, y: 0})
print(f"H at (0, 0) =")
print(H_at_origin)

det_H = H_at_origin.det()
H_11 = H_at_origin[0, 0]

print(f"\nDeterminant of H: {det_H}")
print(f"First element H[0,0]: {H_11}")

if det_H > 0 and H_11 > 0:
    print("\n✓ det(H) > 0 and H[0,0] > 0 → LOCAL MINIMUM")
elif det_H > 0 and H_11 < 0:
    print("\n✓ det(H) > 0 and H[0,0] < 0 → LOCAL MAXIMUM")
elif det_H < 0:
    print("\n✓ det(H) < 0 → SADDLE POINT")
else:
    print("\n? det(H) = 0 → Test is inconclusive")
```

```python
# Visualize the function
fig = plt.figure(figsize=(12, 5))

# 3D plot
ax1 = fig.add_subplot(121, projection='3d')
x_vals = np.linspace(-3, 3, 100)
y_vals = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = X**2 + Y**2

surf = ax1.plot_surface(X, Y, Z, cmap=cm.viridis, alpha=0.8)
ax1.scatter([0], [0], [0], color='red', s=100, marker='o', label='Critical point (0,0)')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('f(x, y)')
ax1.set_title('f(x, y) = x² + y²\n(Local Minimum at origin)')
ax1.legend()

# Contour plot
ax2 = fig.add_subplot(122)
contour = ax2.contour(X, Y, Z, levels=20, cmap=cm.viridis)
ax2.clabel(contour, inline=True, fontsize=8)
ax2.plot(0, 0, 'ro', markersize=10, label='Minimum at (0,0)')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('Contour Plot: Level curves')
ax2.legend()
ax2.grid(True)
ax2.axis('equal')

plt.tight_layout()
plt.show()

print("The bowl shape confirms this is a minimum!")
```

## Example 3: Local Maximum - $f(x, y) = -x^2 - y^2$

Now let's flip the function upside down by adding a negative sign. This creates a paraboloid opening downward (like an upside-down bowl).

```python
# Define function: f(x,y) = -x^2 - y^2
f_max = -x**2 - y**2

print("Function: f(x, y) =", f_max)
print("\n" + "="*50)
print("STEP 1: Find the Jacobian (gradient)")
print("="*50)

# Jacobian
J_max = sp.Matrix([sp.diff(f_max, x), sp.diff(f_max, y)])
print("Jacobian J =", J_max.T)

# Evaluate at (0, 0)
J_at_origin_max = J_max.subs({x: 0, y: 0})
print(f"\nAt point (0, 0): J = {J_at_origin_max.T}")
print("→ Gradient is ZERO! This is a critical point.")

print("\n" + "="*50)
print("STEP 2: Compute the Hessian")
print("="*50)

# Hessian
H_max = sp.hessian(f_max, [x, y])
print("Hessian H =")
print(H_max)
print("\nNote: The Hessian is still constant (doesn't depend on x or y)")

print("\n" + "="*50)
print("STEP 3: Apply the Hessian Test")
print("="*50)

# Evaluate at (0, 0)
H_at_origin_max = H_max.subs({x: 0, y: 0})
print(f"H at (0, 0) =")
print(H_at_origin_max)

det_H_max = H_at_origin_max.det()
H_11_max = H_at_origin_max[0, 0]

print(f"\nDeterminant of H: {det_H_max}")
print(f"First element H[0,0]: {H_11_max}")

if det_H_max > 0 and H_11_max > 0:
    print("\n✓ det(H) > 0 and H[0,0] > 0 → LOCAL MINIMUM")
elif det_H_max > 0 and H_11_max < 0:
    print("\n✓ det(H) > 0 and H[0,0] < 0 → LOCAL MAXIMUM")
elif det_H_max < 0:
    print("\n✓ det(H) < 0 → SADDLE POINT")
else:
    print("\n? det(H) = 0 → Test is inconclusive")
```

```python
# Visualize the function
fig = plt.figure(figsize=(12, 5))

# 3D plot
ax1 = fig.add_subplot(121, projection='3d')
x_vals = np.linspace(-3, 3, 100)
y_vals = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = -X**2 - Y**2

surf = ax1.plot_surface(X, Y, Z, cmap=cm.plasma, alpha=0.8)
ax1.scatter([0], [0], [0], color='red', s=100, marker='o', label='Critical point (0,0)')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('f(x, y)')
ax1.set_title('f(x, y) = -x² - y²\n(Local Maximum at origin)')
ax1.legend()

# Contour plot
ax2 = fig.add_subplot(122)
contour = ax2.contour(X, Y, Z, levels=20, cmap=cm.plasma)
ax2.clabel(contour, inline=True, fontsize=8)
ax2.plot(0, 0, 'ro', markersize=10, label='Maximum at (0,0)')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('Contour Plot: Level curves')
ax2.legend()
ax2.grid(True)
ax2.axis('equal')

plt.tight_layout()
plt.show()

print("The upside-down bowl shape confirms this is a maximum!")
```

## Example 4: Saddle Point - $f(x, y) = x^2 - y^2$

Now for the most interesting case! This function has a **saddle point** at the origin.

### What is a Saddle Point?

A saddle point is a location where:
- The **gradient is zero** (flat in all directions locally)
- But it's **NOT a maximum or minimum**
- The surface curves **up** in some directions and **down** in others

Think of a mountain pass or a saddle on a horse - it's a minimum when you move perpendicular to the ridge, but a maximum when you move along the ridge.

```python
# Define function: f(x,y) = x^2 - y^2 (saddle)
f_saddle = x**2 - y**2

print("Function: f(x, y) =", f_saddle)
print("\n" + "="*50)
print("STEP 1: Find the Jacobian (gradient)")
print("="*50)

# Jacobian
J_saddle = sp.Matrix([sp.diff(f_saddle, x), sp.diff(f_saddle, y)])
print("Jacobian J =", J_saddle.T)

# Evaluate at (0, 0)
J_at_origin_saddle = J_saddle.subs({x: 0, y: 0})
print(f"\nAt point (0, 0): J = {J_at_origin_saddle.T}")
print("→ Gradient is ZERO! This is a critical point.")

print("\n" + "="*50)
print("STEP 2: Compute the Hessian")
print("="*50)

# Hessian
H_saddle = sp.hessian(f_saddle, [x, y])
print("Hessian H =")
print(H_saddle)

print("\n" + "="*50)
print("STEP 3: Apply the Hessian Test")
print("="*50)

# Evaluate at (0, 0)
H_at_origin_saddle = H_saddle.subs({x: 0, y: 0})
print(f"H at (0, 0) =")
print(H_at_origin_saddle)

det_H_saddle = H_at_origin_saddle.det()
H_11_saddle = H_at_origin_saddle[0, 0]

print(f"\nDeterminant of H: {det_H_saddle}")
print(f"First element H[0,0]: {H_11_saddle}")

if det_H_saddle > 0 and H_11_saddle > 0:
    print("\n✓ det(H) > 0 and H[0,0] > 0 → LOCAL MINIMUM")
elif det_H_saddle > 0 and H_11_saddle < 0:
    print("\n✓ det(H) > 0 and H[0,0] < 0 → LOCAL MAXIMUM")
elif det_H_saddle < 0:
    print("\n✓ det(H) < 0 → SADDLE POINT")
    print("   The gradient is flat, but we have slopes coming DOWN in one")
    print("   direction and UP in another direction!")
else:
    print("\n? det(H) = 0 → Test is inconclusive")
```

```python
# Visualize the saddle point
fig = plt.figure(figsize=(12, 5))

# 3D plot
ax1 = fig.add_subplot(121, projection='3d')
x_vals = np.linspace(-3, 3, 100)
y_vals = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = X**2 - Y**2

surf = ax1.plot_surface(X, Y, Z, cmap=cm.coolwarm, alpha=0.8)
ax1.scatter([0], [0], [0], color='red', s=100, marker='o', label='Saddle point (0,0)')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('f(x, y)')
ax1.set_title('f(x, y) = x² - y²\n(Saddle Point at origin)')
ax1.legend()
ax1.view_init(elev=25, azim=45)

# Contour plot
ax2 = fig.add_subplot(122)
contour = ax2.contour(X, Y, Z, levels=30, cmap=cm.coolwarm)
ax2.clabel(contour, inline=True, fontsize=8)
ax2.plot(0, 0, 'ro', markersize=10, label='Saddle point (0,0)')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('Contour Plot: Notice the hyperbolic curves')
ax2.legend()
ax2.grid(True)
ax2.axis('equal')

plt.tight_layout()
plt.show()

print("\nNotice the characteristic 'saddle' or 'Pringle chip' shape!")
print("Along the x-axis (where y=0): function curves UP (like x²)")
print("Along the y-axis (where x=0): function curves DOWN (like -y²)")
```

## Summary: The Hessian Test Algorithm

When you find a critical point (where the gradient/Jacobian = 0), follow these steps:

### Algorithm

1. **Compute the Hessian matrix** $H$ at the critical point
2. **Calculate the determinant** $\det(H)$
3. **Check the first diagonal element** $H_{11}$ (top-left corner)
4. **Apply the rules:**

| Condition | Classification |
|-----------|----------------|
| $\det(H) > 0$ and $H_{11} > 0$ | **Local Minimum** |
| $\det(H) > 0$ and $H_{11} < 0$ | **Local Maximum** |
| $\det(H) < 0$ | **Saddle Point** |
| $\det(H) = 0$ | **Inconclusive** (need other methods) |

### Why This Works (Intuition)

- **Determinant > 0**: The curvatures in all directions have the same sign (all curving up or all curving down)
- **Determinant < 0**: The curvatures have opposite signs (curving up in some directions, down in others)
- **$H_{11}$**: When determinant is positive, this tells us whether we're curving upward (minimum) or downward (maximum)

## Key Properties of the Hessian

### 1. Symmetry
For any continuous function, the Hessian is **symmetric**:
$$H_{ij} = H_{ji}$$

This means once you compute the upper triangular part, you automatically know the lower triangular part!

### 2. Dimension
For a function with $n$ variables, the Hessian is an $n \times n$ square matrix.

### 3. Second-Order Information
While the Jacobian tells us about the **slope** (first derivative), the Hessian tells us about the **curvature** (second derivative).

### 4. Applications in Machine Learning
The Hessian is crucial for:
- **Optimization algorithms** (e.g., Newton's method)
- **Understanding loss landscapes** in neural networks
- **Identifying saddle points** that can trap optimization algorithms
- **Second-order optimization methods** that converge faster than gradient descent

Try this yourself! For the function $f(x, y) = x^2 + xy + y^2$:

1. Find the Jacobian
2. Find the critical point (where Jacobian = 0)
3. Compute the Hessian
4. Classify the critical point

**Solution code below** (try it yourself first!):

```python
# Practice: f(x,y) = x^2 + xy + y^2
f_practice = x**2 + x*y + y**2

print("Function: f(x, y) =", f_practice)
print("\n" + "="*50)
print("Step 1: Jacobian")
print("="*50)

J_practice = sp.Matrix([sp.diff(f_practice, x), sp.diff(f_practice, y)])
print("J =", J_practice.T)

print("\n" + "="*50)
print("Step 2: Find critical point (set J = 0)")
print("="*50)

critical_points = sp.solve(J_practice, [x, y])
print("Critical point:", critical_points)

print("\n" + "="*50)
print("Step 3: Hessian")
print("="*50)

H_practice = sp.hessian(f_practice, [x, y])
print("H =")
print(H_practice)

print("\n" + "="*50)
print("Step 4: Classify the critical point")
print("="*50)

H_at_critical = H_practice.subs(critical_points)
print("H at critical point =")
print(H_at_critical)

det_H_practice = H_at_critical.det()
H_11_practice = H_at_critical[0, 0]

print(f"\ndet(H) = {det_H_practice}")
print(f"H[0,0] = {H_11_practice}")

if det_H_practice > 0 and H_11_practice > 0:
    print("\n✓ LOCAL MINIMUM")
elif det_H_practice > 0 and H_11_practice < 0:
    print("\n✓ LOCAL MAXIMUM")
elif det_H_practice < 0:
    print("\n✓ SADDLE POINT")
else:
    print("\n? INCONCLUSIVE")
```

## Conclusion

The **Hessian matrix** is a powerful extension of the Jacobian that gives us second-order information about multivariable functions:

### What We Learned

1. **Definition**: The Hessian collects all second-order partial derivatives into a symmetric $n \times n$ matrix

2. **Computation Strategy**: 
   - First find the Jacobian (first derivatives)
   - Then differentiate each term again (second derivatives)

3. **Classification of Critical Points**: Using the determinant and first element of the Hessian, we can identify:
   - **Minima** (bowl-shaped)
   - **Maxima** (inverted bowl)
   - **Saddle points** (mountain pass shape)

4. **Key Insight**: Saddle points are particularly important in optimization because they have zero gradient but are neither maxima nor minima. They can cause confusion when searching for optimal points!
