## 1. Understanding Variables and Their Relationships

In single-variable calculus, we often deal with functions like $y = f(x)$, where the relationship is clear. But in real-world problems, the distinction between what's a variable and what's a constant depends heavily on **context**.

### Example: Vehicle Speed and Time

Consider a car's speed as a function of time. In this scenario:
- **Speed** is the **dependent variable** (it depends on time)
- **Time** is the **independent variable** (we can control or observe it independently)

This makes intuitive sense: at any given moment, a car can only have one speed, but the same speed might occur at multiple different times.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import seaborn as sns

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Example: Vehicle speed over time
time = np.linspace(0, 10, 100)
# Simulate a realistic speed profile (acceleration, constant speed, deceleration)
speed = np.piecewise(time, 
                    [time < 3, (time >= 3) & (time < 7), time >= 7],
                    [lambda t: 5 * t,  # acceleration phase
                     lambda t: 15,     # constant speed phase  
                     lambda t: 15 - 2 * (t - 7)])  # deceleration phase

plt.figure(figsize=(10, 6))
plt.plot(time, speed, 'b-', linewidth=2, label='Vehicle Speed')
plt.xlabel('Time (seconds)', fontsize=12)
plt.ylabel('Speed (m/s)', fontsize=12)
plt.title('Vehicle Speed as a Function of Time', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()

# Highlight the concept with annotations
plt.annotate('Acceleration Phase\n(speed depends on time)', 
             xy=(1.5, 7.5), xytext=(1.5, 12),
             arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
             fontsize=10, ha='center', color='red')

plt.annotate('Constant Speed Phase\n(speed independent of time)', 
             xy=(5, 15), xytext=(5, 18),
             arrowprops=dict(arrowstyle='->', color='green', lw=1.5),
             fontsize=10, ha='center', color='green')

plt.tight_layout()
plt.show()

# Demonstrate the one-to-many relationship
print("Key Insight:")
print("• At time t=1s, speed = {:.1f} m/s (one-to-one)".format(np.interp(1, time, speed)))
print("• At time t=5s, speed = {:.1f} m/s (one-to-one)".format(np.interp(5, time, speed)))
print("• But speed = 10 m/s occurs at multiple times (many-to-one)")
print("• This is why time is independent and speed is dependent")
```

## 2. Context Determines Variables vs. Constants

The classification of quantities as variables or constants is **not absolute** - it depends entirely on the context of your problem. Let's explore this with a car engine example.

### Car Engine Force Equation

Consider this simplified relationship for a car's engine:

$$F = ma + dv^2$$

Where:
- $F$ = Force generated by engine
- $m$ = Mass of car  
- $a$ = Acceleration
- $d$ = Aerodynamic drag coefficient
- $v$ = Velocity

### Context 1: You're Driving the Car

When you're behind the wheel:
- **Independent variable**: $F$ (force - you control this with the accelerator)
- **Dependent variables**: $a, v$ (acceleration and velocity change based on your input)
- **Constants**: $m, d$ (mass and drag are fixed features of the car)

### Context 2: You're Designing the Car

As an automotive engineer with performance targets:
- **Independent variable**: $F$ (still the engine force you're designing)
- **Constants**: $a, v$ (fixed performance targets from market research)
- **Design variables (parameters)**: $m, d$ (you can change these by redesigning the car)

This illustrates how the same equation can have different interpretations depending on your perspective!

```python
# Visualize the context-dependent nature of variables
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Context 1: Driver's perspective
force_range = np.linspace(1000, 5000, 50)  # Force in Newtons
m_fixed = 1500  # kg (fixed car mass)
d_fixed = 0.3   # drag coefficient (fixed)

# Calculate resulting acceleration and velocity for different forces
# Simplified: assume v^2 term is small, so F ≈ ma
acceleration = force_range / m_fixed

ax1.plot(force_range, acceleration, 'b-', linewidth=3, label='Acceleration vs Force')
ax1.set_xlabel('Engine Force (N)', fontsize=12)
ax1.set_ylabel('Acceleration (m/s²)', fontsize=12)
ax1.set_title('Driver\'s Perspective:\nForce → Acceleration', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.text(0.05, 0.95, 'Constants:\n• Mass = 1500 kg\n• Drag = 0.3', 
         transform=ax1.transAxes, fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))

# Context 2: Engineer's perspective  
target_acceleration = 3.0  # m/s² (fixed target)
target_force = 4000  # N (fixed engine capability)
mass_range = np.linspace(1000, 2000, 50)  # kg (design variable)

# Calculate required drag coefficient to meet targets
# F = ma + dv^2, assuming some velocity, solve for d
v_typical = 20  # m/s typical velocity
d_required = (target_force - mass_range * target_acceleration) / (v_typical**2)

ax2.plot(mass_range, d_required, 'r-', linewidth=3, label='Required Drag Coefficient')
ax2.set_xlabel('Car Mass (kg)', fontsize=12)
ax2.set_ylabel('Drag Coefficient', fontsize=12)
ax2.set_title('Engineer\'s Perspective:\nMass ↔ Drag Trade-off', fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.text(0.05, 0.95, 'Constants:\n• Acceleration = 3.0 m/s²\n• Force = 4000 N', 
         transform=ax2.transAxes, fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))

plt.tight_layout()
plt.show()

print("The same equation F = ma + dv² has different variable interpretations:")
print("📍 Driver context: F is input, a,v are outputs, m,d are fixed")
print("🔧 Engineer context: a,v are targets, F is capability, m,d are design choices")
```

## 3. Parameters: The Bridge Between Variables and Constants

**Parameters** are quantities that:
- Can be varied to explore different scenarios
- Are often held constant within a single analysis
- Define a "family" of similar functions

### Key Insight About Parameters

When we say something is a parameter, we're typically thinking about:
- **Function fitting**: Finding the best parameter values to match data
- **Sensitivity analysis**: Understanding how changes in parameters affect outcomes  
- **Design optimization**: Choosing parameter values to achieve desired performance

In machine learning and data science, parameters are what we optimize during training (like weights in neural networks or coefficients in regression).

### Mathematical Perspective

From a pure mathematical standpoint, you can differentiate any term with respect to any other term. The labels "variable," "constant," and "parameter" are more about the **context and purpose** of your analysis than mathematical limitations.

> **Important**: Don't be surprised when quantities you thought were constants suddenly become variables in a different context!

## 4. Practical Example: Metal Can Design

Let's work through a concrete example that demonstrates partial derivatives. Imagine you're manufacturing metal cans and need to understand how the can's mass relates to its design parameters.

### The Problem Setup

We want to find the mass $m$ of an empty cylindrical can. The can consists of:
- **Two circular ends** (top and bottom): area = $\pi r^2$ each
- **Cylindrical body**: when "unrolled" becomes a rectangle with width = circumference = $2\pi r$ and height = $h$

### Deriving the Mass Formula

The total volume of metal is:
- Volume = (area of circles + area of rectangle) × thickness
- Volume = $(2\pi r^2 + 2\pi r h) \times t$

The mass is:
- Mass = Volume × Density
- $$m = (2\pi r^2 + 2\pi r h) \times t \times \rho$$

Expanding this:
$$m = 2\pi r^2 t \rho + 2\pi r h t \rho$$

Where:
- $r$ = radius
- $h$ = height  
- $t$ = wall thickness
- $\rho$ = material density
- $\pi$ = definitely a constant! 😊

```python
# Visualize the metal can and its components
fig = plt.figure(figsize=(15, 10))

# 3D visualization of the can
ax1 = fig.add_subplot(2, 2, 1, projection='3d')

# Create cylinder
theta = np.linspace(0, 2*np.pi, 50)
z = np.linspace(0, 1, 50)
r_can = 0.5
h_can = 1.0

# Cylinder surface
theta_mesh, z_mesh = np.meshgrid(theta, z)
x_mesh = r_can * np.cos(theta_mesh)
y_mesh = r_can * np.sin(theta_mesh)

ax1.plot_surface(x_mesh, y_mesh, z_mesh, alpha=0.7, color='lightblue')

# Top and bottom circles
x_circle = r_can * np.cos(theta)
y_circle = r_can * np.sin(theta)
z_bottom = np.zeros_like(theta)
z_top = np.ones_like(theta)

ax1.plot(x_circle, y_circle, z_bottom, 'b-', linewidth=2)
ax1.plot(x_circle, y_circle, z_top, 'b-', linewidth=2)

ax1.set_xlabel('X')
ax1.set_ylabel('Y') 
ax1.set_zlabel('Z')
ax1.set_title('3D View of Can', fontweight='bold')

# 2D "unrolled" view
ax2 = fig.add_subplot(2, 2, 2)

# Draw rectangles representing the unrolled surfaces
# Bottom circle
circle1 = plt.Circle((0.5, 0.5), 0.4, fill=False, color='red', linewidth=2)
ax2.add_patch(circle1)
ax2.text(0.5, 0.5, 'Bottom\n$\\pi r^2$', ha='center', va='center', fontsize=10, color='red')

# Top circle  
circle2 = plt.Circle((2.5, 0.5), 0.4, fill=False, color='red', linewidth=2)
ax2.add_patch(circle2)
ax2.text(2.5, 0.5, 'Top\n$\\pi r^2$', ha='center', va='center', fontsize=10, color='red')

# Unrolled body (rectangle)
rectangle = plt.Rectangle((0.5, 1.5), 2, 1, fill=False, color='blue', linewidth=2)
ax2.add_patch(rectangle)
ax2.text(1.5, 2, 'Unrolled Body\nWidth = $2\\pi r$\nHeight = $h$', 
         ha='center', va='center', fontsize=10, color='blue')

ax2.set_xlim(0, 3)
ax2.set_ylim(0, 3)
ax2.set_aspect('equal')
ax2.set_title('Unrolled Can Components', fontweight='bold')
ax2.axis('off')

# Formula breakdown
ax3 = fig.add_subplot(2, 1, 2)
ax3.text(0.5, 0.8, 'Mass Formula Breakdown:', fontsize=16, fontweight='bold', ha='center')
ax3.text(0.1, 0.6, '• Two circular ends: $2 \\times \\pi r^2 \\times t \\times \\rho = 2\\pi r^2 t \\rho$', 
         fontsize=14, ha='left', color='red')
ax3.text(0.1, 0.4, '• Cylindrical body: $2\\pi r \\times h \\times t \\times \\rho = 2\\pi r h t \\rho$', 
         fontsize=14, ha='left', color='blue') 
ax3.text(0.1, 0.2, '• Total mass: $m = 2\\pi r^2 t \\rho + 2\\pi r h t \\rho$', 
         fontsize=16, ha='left', fontweight='bold', color='black')

ax3.set_xlim(0, 1)
ax3.set_ylim(0, 1)
ax3.axis('off')

plt.tight_layout()
plt.show()

# Show some example calculations
print("Example Can Specifications:")
print("• Radius (r) = 5 cm")
print("• Height (h) = 10 cm") 
print("• Thickness (t) = 0.1 cm")
print("• Density (ρ) = 2.7 g/cm³ (aluminum)")
print()

r, h, t, rho = 5, 10, 0.1, 2.7
mass = 2*np.pi*r**2*t*rho + 2*np.pi*r*h*t*rho
print(f"Calculated mass = {mass:.1f} grams")
```

## 5. Introduction to Partial Derivatives

Now comes the key concept: **partial derivatives**. When we have a function of multiple variables, we can find how the function changes with respect to one variable while **keeping all others constant**.

### The Partial Derivative Symbol

Instead of using $\frac{d}{dx}$ (ordinary derivative), we use $\frac{\partial}{\partial x}$ (partial derivative) to indicate we're working with multivariable functions.

### The Process

To compute $\frac{\partial f}{\partial x}$:
1. Treat all variables except $x$ as constants
2. Apply the same differentiation rules you know from single-variable calculus
3. Differentiate only the terms containing $x$

### Why This Works

This approach essentially "flattens" a multidimensional problem into a 1D problem by focusing on one variable at a time. It's like taking a cross-section of a 3D surface to see how it slopes in one direction.

Let's apply this to our can mass formula:
$$m = 2\pi r^2 t \rho + 2\pi r h t \rho$$

```python
# Let's calculate all partial derivatives step by step
import sympy as sp

print("PARTIAL DERIVATIVES OF CAN MASS FUNCTION")
print("="*50)
print("Given: m = 2πr²tρ + 2πrhtρ")
print()

# Define symbols
r, h, t, rho = sp.symbols('r h t rho', positive=True)
pi = sp.pi

# Define the mass function
m = 2*pi*r**2*t*rho + 2*pi*r*h*t*rho

print("Original function:")
print(f"m = {m}")
print()

# Calculate partial derivatives
print("CALCULATING PARTIAL DERIVATIVES:")
print("-" * 30)

# 1. Partial derivative with respect to h
dm_dh = sp.diff(m, h)
print("1. ∂m/∂h (treating r, t, ρ as constants):")
print(f"   First term: 2πr²tρ → 0 (no h term)")
print(f"   Second term: 2πrhtρ → 2πrtρ (h differentiates to 1)")
print(f"   Result: ∂m/∂h = {dm_dh}")
print(f"   Interpretation: Mass changes linearly with height")
print()

# 2. Partial derivative with respect to r  
dm_dr = sp.diff(m, r)
print("2. ∂m/∂r (treating h, t, ρ as constants):")
print(f"   First term: 2πr²tρ → 4πrtρ (r² differentiates to 2r)")
print(f"   Second term: 2πrhtρ → 2πhtρ (r differentiates to 1)")
print(f"   Result: ∂m/∂r = {dm_dr}")
print(f"   Interpretation: Mass changes nonlinearly with radius")
print()

# 3. Partial derivative with respect to t
dm_dt = sp.diff(m, t) 
print("3. ∂m/∂t (treating r, h, ρ as constants):")
print(f"   Both terms contain t as a simple factor")
print(f"   Result: ∂m/∂t = {dm_dt}")
print(f"   Interpretation: Mass changes linearly with thickness")
print()

# 4. Partial derivative with respect to rho
dm_drho = sp.diff(m, rho)
print("4. ∂m/∂ρ (treating r, h, t as constants):")
print(f"   Both terms contain ρ as a simple factor")
print(f"   Result: ∂m/∂ρ = {dm_drho}")
print(f"   Interpretation: Mass changes linearly with density")
print()

print("KEY OBSERVATIONS:")
print("• Only ∂m/∂r contains the variable r (nonlinear relationship)")
print("• All other partial derivatives are constants or don't contain their variable")
print("• This tells us how sensitive the mass is to changes in each parameter")
```

```python
# Visualize how mass changes with each parameter
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Base parameters
r_base, h_base, t_base, rho_base = 5, 10, 0.1, 2.7

# 1. Mass vs Height (linear relationship)
h_range = np.linspace(5, 20, 50)
mass_vs_h = 2*np.pi*r_base**2*t_base*rho_base + 2*np.pi*r_base*h_range*t_base*rho_base

axes[0,0].plot(h_range, mass_vs_h, 'b-', linewidth=3)
axes[0,0].set_xlabel('Height (cm)')
axes[0,0].set_ylabel('Mass (g)')
axes[0,0].set_title('Mass vs Height\n(Linear relationship)', fontweight='bold')
axes[0,0].grid(True, alpha=0.3)
axes[0,0].text(0.05, 0.95, f'∂m/∂h = {2*np.pi*r_base*t_base*rho_base:.1f}', 
               transform=axes[0,0].transAxes, fontsize=12, 
               bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))

# 2. Mass vs Radius (nonlinear relationship)  
r_range = np.linspace(2, 8, 50)
mass_vs_r = 2*np.pi*r_range**2*t_base*rho_base + 2*np.pi*r_range*h_base*t_base*rho_base

axes[0,1].plot(r_range, mass_vs_r, 'r-', linewidth=3)
axes[0,1].set_xlabel('Radius (cm)')
axes[0,1].set_ylabel('Mass (g)')
axes[0,1].set_title('Mass vs Radius\n(Nonlinear relationship)', fontweight='bold')
axes[0,1].grid(True, alpha=0.3)
axes[0,1].text(0.05, 0.95, '∂m/∂r = 4πrtρ + 2πhtρ', 
               transform=axes[0,1].transAxes, fontsize=10,
               bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))

# 3. Mass vs Thickness (linear relationship)
t_range = np.linspace(0.05, 0.3, 50)  
mass_vs_t = 2*np.pi*r_base**2*t_range*rho_base + 2*np.pi*r_base*h_base*t_range*rho_base

axes[1,0].plot(t_range, mass_vs_t, 'g-', linewidth=3)
axes[1,0].set_xlabel('Thickness (cm)')
axes[1,0].set_ylabel('Mass (g)')
axes[1,0].set_title('Mass vs Thickness\n(Linear relationship)', fontweight='bold')
axes[1,0].grid(True, alpha=0.3)
axes[1,0].text(0.05, 0.95, f'∂m/∂t = {2*np.pi*(r_base**2 + r_base*h_base)*rho_base:.1f}', 
               transform=axes[1,0].transAxes, fontsize=10,
               bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))

# 4. Mass vs Density (linear relationship)
rho_range = np.linspace(1, 5, 50)  # Different materials
mass_vs_rho = 2*np.pi*r_base**2*t_base*rho_range + 2*np.pi*r_base*h_base*t_base*rho_range

axes[1,1].plot(rho_range, mass_vs_rho, 'm-', linewidth=3)
axes[1,1].set_xlabel('Density (g/cm³)')
axes[1,1].set_ylabel('Mass (g)')
axes[1,1].set_title('Mass vs Density\n(Linear relationship)', fontweight='bold')
axes[1,1].grid(True, alpha=0.3)
axes[1,1].text(0.05, 0.95, f'∂m/∂ρ = {2*np.pi*(r_base**2*t_base + r_base*h_base*t_base):.1f}', 
               transform=axes[1,1].transAxes, fontsize=10,
               bbox=dict(boxstyle='round', facecolor='plum', alpha=0.8))

plt.tight_layout()
plt.show()

print("INTERPRETATION OF THE PLOTS:")
print("• Height, thickness, and density show LINEAR relationships with mass")
print("• Radius shows a NONLINEAR (quadratic) relationship with mass")  
print("• The slope of each plot represents the partial derivative")
print("• Steeper slopes mean the mass is more sensitive to that parameter")
```

## 6. Geometric Interpretation of Partial Derivatives

Partial derivatives have a beautiful geometric interpretation. For a function $f(x,y)$:

- $\frac{\partial f}{\partial x}$ represents the **slope** of the function when we move in the $x$-direction while keeping $y$ constant
- $\frac{\partial f}{\partial y}$ represents the **slope** of the function when we move in the $y$-direction while keeping $x$ constant

Think of it as taking "cross-sections" of a 3D surface and measuring how steep those cross-sections are.

### Visualizing This Concept

Let's create a 3D surface and see how partial derivatives work geometrically.

```python
# Create a 3D visualization of partial derivatives
fig = plt.figure(figsize=(16, 12))

# Create a simple 2-variable function for visualization
x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50) 
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2  # Simple paraboloid: f(x,y) = x² + y²

# Main 3D surface plot
ax1 = fig.add_subplot(2, 2, 1, projection='3d')
surf = ax1.plot_surface(X, Y, Z, alpha=0.7, cmap='viridis')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('f(x,y) = x² + y²')
ax1.set_title('3D Surface: f(x,y) = x² + y²', fontweight='bold')

# Show cross-sections at specific points
x_fix = 1.0  # Fixed x value
y_fix = 1.0  # Fixed y value

# Cross-section in x-direction (y fixed)
y_cross = np.full_like(x, y_fix)
z_cross_x = x**2 + y_fix**2
ax1.plot(x, y_cross, z_cross_x, 'r-', linewidth=4, label=f'y = {y_fix} (fixed)')

# Cross-section in y-direction (x fixed)  
x_cross = np.full_like(y, x_fix)
z_cross_y = x_fix**2 + y**2
ax1.plot(x_cross, y, z_cross_y, 'b-', linewidth=4, label=f'x = {x_fix} (fixed)')

ax1.legend()

# Plot the cross-section in x-direction
ax2 = fig.add_subplot(2, 2, 2)
ax2.plot(x, z_cross_x, 'r-', linewidth=3)
ax2.set_xlabel('x')
ax2.set_ylabel('f(x, y=1)')
ax2.set_title('Cross-section: y = 1 (fixed)\n∂f/∂x = 2x', fontweight='bold')
ax2.grid(True, alpha=0.3)

# Show tangent line at a point
x_point = 1.0
z_point = x_point**2 + y_fix**2
slope_x = 2 * x_point  # ∂f/∂x = 2x at this point
x_tangent = np.linspace(x_point - 0.5, x_point + 0.5, 10)
z_tangent = z_point + slope_x * (x_tangent - x_point)
ax2.plot(x_tangent, z_tangent, 'k--', linewidth=2, label=f'Tangent (slope = {slope_x})')
ax2.plot(x_point, z_point, 'ko', markersize=8, label=f'Point ({x_point}, {z_point})')
ax2.legend()

# Plot the cross-section in y-direction
ax3 = fig.add_subplot(2, 2, 3)
ax3.plot(y, z_cross_y, 'b-', linewidth=3)
ax3.set_xlabel('y')
ax3.set_ylabel('f(x=1, y)')
ax3.set_title('Cross-section: x = 1 (fixed)\n∂f/∂y = 2y', fontweight='bold')
ax3.grid(True, alpha=0.3)

# Show tangent line at a point
y_point = 1.0
z_point_y = x_fix**2 + y_point**2
slope_y = 2 * y_point  # ∂f/∂y = 2y at this point
y_tangent = np.linspace(y_point - 0.5, y_point + 0.5, 10)
z_tangent_y = z_point_y + slope_y * (y_tangent - y_point)
ax3.plot(y_tangent, z_tangent_y, 'k--', linewidth=2, label=f'Tangent (slope = {slope_y})')
ax3.plot(y_point, z_point_y, 'ko', markersize=8, label=f'Point ({y_point}, {z_point_y})')
ax3.legend()

# Contour plot showing level curves
ax4 = fig.add_subplot(2, 2, 4)
contour = ax4.contour(X, Y, Z, levels=15, colors='black', alpha=0.6)
ax4.clabel(contour, inline=True, fontsize=8)
contourf = ax4.contourf(X, Y, Z, levels=15, cmap='viridis', alpha=0.8)
ax4.set_xlabel('x')
ax4.set_ylabel('y')
ax4.set_title('Contour Plot\n(Level curves of constant f)', fontweight='bold')

# Show the cross-section lines on contour plot
ax4.axhline(y=y_fix, color='red', linewidth=3, alpha=0.8, label=f'y = {y_fix}')
ax4.axvline(x=x_fix, color='blue', linewidth=3, alpha=0.8, label=f'x = {x_fix}')
ax4.plot(x_fix, y_fix, 'ko', markersize=10, label=f'Point ({x_fix}, {y_fix})')
ax4.legend()

plt.tight_layout()
plt.show()

print("GEOMETRIC INTERPRETATION:")
print("🔴 Red line: Cross-section with y fixed → shows ∂f/∂x")
print("🔵 Blue line: Cross-section with x fixed → shows ∂f/∂y") 
print("⚫ Black dashed: Tangent lines showing the slopes (partial derivatives)")
print("📊 Bottom right: Contour plot shows 'bird's eye view' of the surface")
print()
print("Key insight: Partial derivatives are just slopes of cross-sections!")
```

```python
# Solution to the practice exercise
import sympy as sp

print("PRACTICE EXERCISE SOLUTION")
print("=" * 40)
print("Function: T(x,y,t) = 20 + 5e^(-0.1(x²+y²))sin(0.5t)")
print()

# Define symbols
x, y, t = sp.symbols('x y t', real=True)

# Define the temperature function
T = 20 + 5 * sp.exp(-0.1*(x**2 + y**2)) * sp.sin(0.5*t)

print("Calculating partial derivatives:")
print("-" * 30)

# Partial derivative with respect to x
dT_dx = sp.diff(T, x)
print("∂T/∂x =", dT_dx)
print("Physical meaning: Rate of temperature change in x-direction")
print("                 (temperature gradient in x-direction)")
print()

# Partial derivative with respect to y  
dT_dy = sp.diff(T, y)
print("∂T/∂y =", dT_dy)
print("Physical meaning: Rate of temperature change in y-direction")
print("                 (temperature gradient in y-direction)")
print()

# Partial derivative with respect to t
dT_dt = sp.diff(T, t)
print("∂T/∂t =", dT_dt)
print("Physical meaning: Rate of temperature change over time")
print("                 (how fast temperature is changing at each point)")
print()

# Create visualization
fig = plt.figure(figsize=(16, 10))

# Parameters for visualization
x_vals = np.linspace(-5, 5, 50)
y_vals = np.linspace(-5, 5, 50)
t_vals = np.linspace(0, 4*np.pi, 100)

X, Y = np.meshgrid(x_vals, y_vals, indexing='xy')

# Temperature at different times
times = [0, np.pi/2, np.pi, 3*np.pi/2]
time_labels = ['t = 0', 't = π/2', 't = π', 't = 3π/2']

for i, (time_val, label) in enumerate(zip(times, time_labels)):
    ax = fig.add_subplot(2, 2, i+1)
    
    # Calculate temperature at this time
    T_vals = 20 + 5 * np.exp(-0.1*(X**2 + Y**2)) * np.sin(0.5*time_val)
    
    # Create contour plot
    contour = ax.contourf(X, Y, T_vals, levels=20, cmap='coolwarm')
    plt.colorbar(contour, ax=ax, label='Temperature (°C)')
    
    # Add contour lines
    contour_lines = ax.contour(X, Y, T_vals, levels=10, colors='black', alpha=0.4, linewidths=0.5)
    
    ax.set_xlabel('x position')
    ax.set_ylabel('y position') 
    ax.set_title(f'Temperature Distribution at {label}', fontweight='bold')
    ax.set_aspect('equal')

plt.tight_layout()
plt.show()

# Show temperature evolution at the center
plt.figure(figsize=(12, 4))
T_center = 20 + 5 * np.sin(0.5*t_vals)
plt.plot(t_vals, T_center, 'b-', linewidth=3, label='Temperature at center (0,0)')
plt.axhline(y=20, color='k', linestyle='--', alpha=0.5, label='Base temperature')
plt.xlabel('Time')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Oscillation at Center Over Time', fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("\nKEY INSIGHTS:")
print("• ∂T/∂x and ∂T/∂y show how temperature varies spatially")
print("• ∂T/∂t shows how temperature varies temporally")
print("• All three partial derivatives work together to describe the full behavior")
print("• This is typical in physics: space and time derivatives describe different aspects")
```

## 8. Summary and Key Takeaways

### 🎯 What We've Learned

1. **Context is Everything**
   - Variables, constants, and parameters are **contextual labels**
   - The same quantity can be a variable in one problem and a constant in another
   - Always ask: "What am I trying to understand or optimize?"

2. **Partial Derivatives are Intuitive**
   - Just treat other variables as constants and differentiate normally
   - Use $\frac{\partial}{\partial x}$ notation for multivariable functions
   - Geometrically: they represent slopes of cross-sections

3. **Physical Interpretation Matters**
   - Partial derivatives often have clear physical meanings
   - $\frac{\partial T}{\partial x}$ = spatial temperature gradient
   - $\frac{\partial T}{\partial t}$ = rate of temperature change over time
   - $\frac{\partial m}{\partial r}$ = how mass changes with radius

4. **Practical Applications**
   - Engineering design optimization
   - Physics and heat transfer
   - Machine learning (gradient descent!)
   - Economics (marginal effects)
