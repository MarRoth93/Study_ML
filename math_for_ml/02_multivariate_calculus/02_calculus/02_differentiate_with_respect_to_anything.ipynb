## 1. Partial Derivatives of Complex Functions

In the previous notebook, we worked with relatively simple functions. Now let's tackle more complex examples that involve combinations of trigonometric, exponential, and polynomial terms.

### Example Function
Consider this function of three variables:
$$f(x, y, z) = \sin(x) \cdot e^{yz^2}$$

This function combines:
- A **trigonometric term**: $\sin(x)$
- An **exponential term**: $e^{yz^2}$
- **Three variables**: $x$, $y$, and $z$

Let's find all three partial derivatives step by step.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import sympy as sp
import seaborn as sns
from matplotlib.patches import FancyBboxPatch

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Configure matplotlib for better display
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 11

print("📚 Ready to explore partial derivatives and total derivatives!")
print("🎯 Today's function: f(x,y,z) = sin(x) · e^(yz²)")
```

## 2. Step-by-Step Partial Derivative Calculations

Let's calculate each partial derivative systematically. Remember the key principle:
> **When differentiating with respect to one variable, treat all other variables as constants.**

### The Function
$$f(x, y, z) = \sin(x) \cdot e^{yz^2}$$

### Strategy for Each Variable
- **For $\frac{\partial f}{\partial x}$**: Treat $y$ and $z$ as constants
- **For $\frac{\partial f}{\partial y}$**: Treat $x$ and $z$ as constants  
- **For $\frac{\partial f}{\partial z}$**: Treat $x$ and $y$ as constants

```python
# Define symbols and the function
x, y, z = sp.symbols('x y z', real=True)
f = sp.sin(x) * sp.exp(y * z**2)

print("CALCULATING PARTIAL DERIVATIVES")
print("=" * 50)
print(f"Original function: f(x,y,z) = {f}")
print()

# 1. Partial derivative with respect to x
print("1️⃣ PARTIAL DERIVATIVE WITH RESPECT TO x")
print("-" * 40)
print("f(x,y,z) = sin(x) · e^(yz²)")
print("         = sin(x) · [constant w.r.t. x]")
print()
print("Apply product rule: d/dx[sin(x) · constant] = cos(x) · constant")

df_dx = sp.diff(f, x)
print(f"∂f/∂x = {df_dx}")
print("✅ The exponential term e^(yz²) doesn't contain x, so it's treated as constant")
print("✅ d/dx[sin(x)] = cos(x)")
print()

# 2. Partial derivative with respect to y  
print("2️⃣ PARTIAL DERIVATIVE WITH RESPECT TO y")
print("-" * 40)
print("f(x,y,z) = sin(x) · e^(yz²)")
print("         = [constant w.r.t. y] · e^(yz²)")
print()
print("Apply product rule and chain rule:")
print("d/dy[constant · e^(yz²)] = constant · e^(yz²) · d/dy[yz²]")
print("                        = constant · e^(yz²) · z²")

df_dy = sp.diff(f, y)
print(f"∂f/∂y = {df_dy}")
print("✅ sin(x) is constant w.r.t. y") 
print("✅ Chain rule: d/dy[e^(yz²)] = e^(yz²) · d/dy[yz²] = e^(yz²) · z²")
print()

# 3. Partial derivative with respect to z
print("3️⃣ PARTIAL DERIVATIVE WITH RESPECT TO z")
print("-" * 40)
print("f(x,y,z) = sin(x) · e^(yz²)")
print("         = [constant w.r.t. z] · e^(yz²)")
print()
print("Apply product rule and chain rule:")
print("d/dz[constant · e^(yz²)] = constant · e^(yz²) · d/dz[yz²]")
print("                        = constant · e^(yz²) · 2yz")

df_dz = sp.diff(f, z)
print(f"∂f/∂z = {df_dz}")
print("✅ sin(x) is constant w.r.t. z")
print("✅ Chain rule: d/dz[e^(yz²)] = e^(yz²) · d/dz[yz²] = e^(yz²) · 2yz")
print()

print("🎯 SUMMARY OF ALL PARTIAL DERIVATIVES:")
print("=" * 50)
print(f"∂f/∂x = {df_dx}")
print(f"∂f/∂y = {df_dy}") 
print(f"∂f/∂z = {df_dz}")
```

```python
# Visualize the function and its behavior
fig = plt.figure(figsize=(16, 12))

# Create a grid for visualization
x_vals = np.linspace(-np.pi, np.pi, 50)
y_vals = np.linspace(-1, 1, 50)
z_val = 0.5  # Fixed z value for visualization

X, Y = np.meshgrid(x_vals, y_vals)
Z_fixed = z_val * np.ones_like(X)

# Calculate function values
F_vals = np.sin(X) * np.exp(Y * z_val**2)

# Calculate partial derivative values
dF_dx_vals = np.cos(X) * np.exp(Y * z_val**2)
dF_dy_vals = np.sin(X) * np.exp(Y * z_val**2) * z_val**2
dF_dz_vals = np.sin(X) * np.exp(Y * z_val**2) * 2 * Y * z_val

# Plot 1: Original function
ax1 = fig.add_subplot(2, 2, 1, projection='3d')
surf1 = ax1.plot_surface(X, Y, F_vals, cmap='viridis', alpha=0.8)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('f(x,y,z=0.5)')
ax1.set_title('Original Function\nf(x,y,z) = sin(x)·e^(yz²)', fontweight='bold')

# Plot 2: ∂f/∂x
ax2 = fig.add_subplot(2, 2, 2, projection='3d')
surf2 = ax2.plot_surface(X, Y, dF_dx_vals, cmap='coolwarm', alpha=0.8)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('∂f/∂x')
ax2.set_title('Partial Derivative w.r.t. x\n∂f/∂x = cos(x)·e^(yz²)', fontweight='bold')

# Plot 3: ∂f/∂y  
ax3 = fig.add_subplot(2, 2, 3, projection='3d')
surf3 = ax3.plot_surface(X, Y, dF_dy_vals, cmap='plasma', alpha=0.8)
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_zlabel('∂f/∂y')
ax3.set_title('Partial Derivative w.r.t. y\n∂f/∂y = sin(x)·e^(yz²)·z²', fontweight='bold')

# Plot 4: Cross-sections showing behavior
ax4 = fig.add_subplot(2, 2, 4)

# Cross-section at y=0.5
y_cross = 0.5
f_cross = np.sin(x_vals) * np.exp(y_cross * z_val**2)
df_dx_cross = np.cos(x_vals) * np.exp(y_cross * z_val**2)

ax4.plot(x_vals, f_cross, 'b-', linewidth=3, label='f(x, y=0.5, z=0.5)')
ax4.plot(x_vals, df_dx_cross, 'r--', linewidth=2, label='∂f/∂x at y=0.5, z=0.5')
ax4.set_xlabel('x')
ax4.set_ylabel('Function Value')
ax4.set_title('Cross-sections: Function vs Its Derivative', fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("🔍 VISUALIZATION INSIGHTS:")
print("• Top-left: Original function combines oscillation (sin) with exponential growth")
print("• Top-right: ∂f/∂x has the same exponential shape but with cosine oscillation")
print("• Bottom-left: ∂f/∂y amplifies the original function by z² factor")
print("• Bottom-right: Cross-section shows how derivative relates to function slope")
```

## 3. The Total Derivative: A Powerful Extension

Now we introduce one of the most important concepts in multivariable calculus: the **total derivative**.

### The Scenario
Imagine that our variables $x$, $y$, and $z$ are not independent, but are themselves functions of another parameter $t$:

$$\begin{align}
x(t) &= t - 1 \\
y(t) &= t^2 \\
z(t) &= \frac{1}{t}
\end{align}$$

**Question**: How does $f$ change with respect to $t$?

### Two Approaches

#### Approach 1: Direct Substitution (Brute Force)
We could substitute the expressions for $x(t)$, $y(t)$, and $z(t)$ directly into $f$:
$$f(t) = \sin(t-1) \cdot e^{t^2 \cdot (1/t)^2} = \sin(t-1) \cdot e^{1}$$

Then differentiate: $\frac{df}{dt} = \cos(t-1) \cdot e$

#### Approach 2: Total Derivative (Chain Rule)
For complex functions, direct substitution becomes unwieldy. The **total derivative** uses the chain rule:

$$\frac{df}{dt} = \frac{\partial f}{\partial x}\frac{dx}{dt} + \frac{\partial f}{\partial y}\frac{dy}{dt} + \frac{\partial f}{\partial z}\frac{dz}{dt}$$

This is the **multivariable chain rule**!

```python
# Calculate the total derivative step by step
print("CALCULATING THE TOTAL DERIVATIVE")
print("=" * 50)

# Define the parameter t and the functions x(t), y(t), z(t)
t = sp.symbols('t', real=True, positive=True)
x_t = t - 1
y_t = t**2
z_t = 1/t

print(f"Given parametric functions:")
print(f"x(t) = {x_t}")
print(f"y(t) = {y_t}")
print(f"z(t) = {z_t}")
print()

# Calculate derivatives of parametric functions
dx_dt = sp.diff(x_t, t)
dy_dt = sp.diff(y_t, t)  
dz_dt = sp.diff(z_t, t)

print("Step 1: Find derivatives of parametric functions")
print("-" * 30)
print(f"dx/dt = {dx_dt}")
print(f"dy/dt = {dy_dt}")
print(f"dz/dt = {dz_dt}")
print()

print("Step 2: Apply the total derivative formula")
print("-" * 30)
print("df/dt = (∂f/∂x)(dx/dt) + (∂f/∂y)(dy/dt) + (∂f/∂z)(dz/dt)")
print()

# We already calculated the partial derivatives
print("Recall our partial derivatives:")
print(f"∂f/∂x = {df_dx}")
print(f"∂f/∂y = {df_dy}")
print(f"∂f/∂z = {df_dz}")
print()

# Calculate total derivative symbolically
total_derivative = df_dx * dx_dt + df_dy * dy_dt + df_dz * dz_dt

print("Step 3: Substitute and multiply")
print("-" * 30)
print("df/dt = cos(x)·e^(yz²)·(1) + sin(x)·e^(yz²)·z²·(2t) + sin(x)·e^(yz²)·(2yz)·(-1/t²)")
print()
print(f"Before simplification:")
print(f"df/dt = {total_derivative}")
print()

# Substitute the parametric functions and simplify
total_derivative_substituted = total_derivative.subs([(x, x_t), (y, y_t), (z, z_t)])
total_derivative_simplified = sp.simplify(total_derivative_substituted)

print("Step 4: Substitute x(t), y(t), z(t) and simplify")
print("-" * 30)
print(f"After substitution: {total_derivative_substituted}")
print(f"After simplification: {total_derivative_simplified}")
print()

# Verify with direct approach
print("🔍 VERIFICATION: Direct substitution approach")
print("-" * 40)
f_direct = sp.sin(x_t) * sp.exp(y_t * z_t**2)
f_direct_simplified = sp.simplify(f_direct)
df_dt_direct = sp.diff(f_direct_simplified, t)

print(f"f(t) = sin(t-1) · e^(t² · (1/t)²)")
print(f"f(t) = sin(t-1) · e^(t² · 1/t²)")  
print(f"f(t) = sin(t-1) · e^1")
print(f"f(t) = {f_direct_simplified}")
print()
print(f"df/dt (direct) = {df_dt_direct}")
print(f"df/dt (total)  = {total_derivative_simplified}")
print()

# Check if they're equal
are_equal = sp.simplify(df_dt_direct - total_derivative_simplified) == 0
print(f"✅ Results match: {are_equal}")

if are_equal:
    print("🎉 Both approaches give the same result!")
else:
    print("⚠️  Let's check the algebra again...")
```

```python
# Let's examine the "miraculous" cancellation mentioned in the transcript
print("🔬 DETAILED ANALYSIS: The Miraculous Cancellation")
print("=" * 60)

# Let's substitute step by step to see the cancellation
print("Starting with the total derivative formula:")
print("df/dt = (∂f/∂x)(dx/dt) + (∂f/∂y)(dy/dt) + (∂f/∂z)(dz/dt)")
print()

# Substitute the values we know
print("Substituting our known values:")
print("df/dt = cos(x)·e^(yz²)·(1) + sin(x)·e^(yz²)·z²·(2t) + sin(x)·e^(yz²)·(2yz)·(-1/t²)")
print()

# Now substitute x(t), y(t), z(t)
print("Substituting x = t-1, y = t², z = 1/t:")
print("df/dt = cos(t-1)·e^(t²·(1/t)²)·(1) + sin(t-1)·e^(t²·(1/t)²)·(1/t)²·(2t) + sin(t-1)·e^(t²·(1/t)²)·(2·t²·1/t)·(-1/t²)")
print()

print("Simplifying the exponent: t²·(1/t)² = t²·(1/t²) = 1")
print("So e^(t²·(1/t)²) = e^1 = e")
print()

print("Our three terms become:")
print("Term 1: cos(t-1)·e·(1) = e·cos(t-1)")
print("Term 2: sin(t-1)·e·(1/t²)·(2t) = sin(t-1)·e·(2t/t²) = sin(t-1)·e·(2/t)")
print("Term 3: sin(t-1)·e·(2t²/t)·(-1/t²) = sin(t-1)·e·(2t)·(-1/t²) = sin(t-1)·e·(-2/t)")
print()

print("🎯 Notice something amazing:")
print("Term 2: +sin(t-1)·e·(2/t)")
print("Term 3: -sin(t-1)·e·(2/t)")
print("These are EXACTLY OPPOSITE and cancel out!")
print()

print("Final result:")
print("df/dt = e·cos(t-1) + [sin(t-1)·e·(2/t)] + [sin(t-1)·e·(-2/t)]")
print("df/dt = e·cos(t-1) + 0")
print("df/dt = e·cos(t-1)")
print()

# Verify numerically at a specific point
t_val = 2.0
x_val = t_val - 1  # = 1
y_val = t_val**2   # = 4  
z_val = 1/t_val    # = 0.5

print(f"🧮 NUMERICAL VERIFICATION at t = {t_val}:")
print(f"x({t_val}) = {x_val}, y({t_val}) = {y_val}, z({t_val}) = {z_val}")
print()

# Calculate each term numerically
term1 = np.cos(x_val) * np.exp(y_val * z_val**2) * 1
term2 = np.sin(x_val) * np.exp(y_val * z_val**2) * z_val**2 * (2*t_val)
term3 = np.sin(x_val) * np.exp(y_val * z_val**2) * (2*y_val*z_val) * (-1/t_val**2)

total_numerical = term1 + term2 + term3
direct_numerical = np.exp(1) * np.cos(t_val - 1)

print(f"Term 1: {term1:.6f}")
print(f"Term 2: {term2:.6f}")
print(f"Term 3: {term3:.6f}")
print(f"Sum of terms 2 and 3: {term2 + term3:.10f}")
print()
print(f"Total derivative (sum): {total_numerical:.6f}")
print(f"Direct approach:       {direct_numerical:.6f}")
print(f"Difference:            {abs(total_numerical - direct_numerical):.2e}")
print()
print("✅ Perfect agreement! The cancellation works exactly as predicted.")
```

## 4. Why the Total Derivative is So Powerful

The total derivative might seem like unnecessary complexity, but it's actually incredibly useful in many scenarios:

### 🔹 When Direct Substitution is Impractical

1. **Complex expressions**: When substituting creates unwieldy expressions
2. **No analytical form**: When we don't have explicit formulas for $x(t)$, $y(t)$, $z(t)$
3. **Numerical data**: When variables are given as data points rather than functions
4. **Computational efficiency**: When partial derivatives are already computed

### 🔹 Real-World Applications

- **Physics**: Temperature changing along a moving particle's path
- **Engineering**: Stress in a material as geometry changes over time  
- **Economics**: Utility function with changing preference parameters
- **Machine Learning**: Loss function during gradient descent optimization

### 🔹 The Chain Rule Pattern

The total derivative is the **multivariable version of the chain rule**:

**Single variable**: $\frac{df}{dt} = f'(g(t)) \cdot g'(t)$

**Multiple variables**: $\frac{df}{dt} = \sum_i \frac{\partial f}{\partial x_i} \frac{dx_i}{dt}$

```python
# Visualize the difference between direct and total derivative approaches
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Generate t values
t_vals = np.linspace(0.5, 3, 100)

# Calculate function values using both approaches
f_direct_vals = []
df_dt_direct_vals = []
df_dt_total_vals = []

for t_val in t_vals:
    # Direct approach: f(t) = sin(t-1) * e
    f_val = np.sin(t_val - 1) * np.exp(1)
    df_val = np.cos(t_val - 1) * np.exp(1)
    
    f_direct_vals.append(f_val)
    df_dt_direct_vals.append(df_val)
    
    # Total derivative approach (should give same result)
    x_val = t_val - 1
    y_val = t_val**2
    z_val = 1/t_val
    
    # Calculate each term
    term1 = np.cos(x_val) * np.exp(y_val * z_val**2) * 1
    term2 = np.sin(x_val) * np.exp(y_val * z_val**2) * z_val**2 * (2*t_val)  
    term3 = np.sin(x_val) * np.exp(y_val * z_val**2) * (2*y_val*z_val) * (-1/t_val**2)
    
    df_dt_total_vals.append(term1 + term2 + term3)

f_direct_vals = np.array(f_direct_vals)
df_dt_direct_vals = np.array(df_dt_direct_vals)
df_dt_total_vals = np.array(df_dt_total_vals)

# Plot 1: Function values
axes[0,0].plot(t_vals, f_direct_vals, 'b-', linewidth=3, label='f(t) = sin(t-1)·e')
axes[0,0].set_xlabel('t')
axes[0,0].set_ylabel('f(t)')
axes[0,0].set_title('Function Values Over Time', fontweight='bold')
axes[0,0].grid(True, alpha=0.3)
axes[0,0].legend()

# Plot 2: Derivative comparison
axes[0,1].plot(t_vals, df_dt_direct_vals, 'r-', linewidth=3, label='Direct approach')
axes[0,1].plot(t_vals, df_dt_total_vals, 'g--', linewidth=2, alpha=0.8, label='Total derivative')
axes[0,1].set_xlabel('t')
axes[0,1].set_ylabel('df/dt')
axes[0,1].set_title('Derivative Comparison', fontweight='bold')
axes[0,1].grid(True, alpha=0.3)
axes[0,1].legend()

# Plot 3: Individual terms in total derivative
axes[1,0].plot(t_vals, [np.cos(t-1)*np.exp(1) for t in t_vals], 'b-', linewidth=2, label='Term 1: cos(t-1)·e')

# Calculate terms 2 and 3 separately to show cancellation
term2_vals = []
term3_vals = []
for t_val in t_vals:
    x_val, y_val, z_val = t_val-1, t_val**2, 1/t_val
    term2 = np.sin(x_val) * np.exp(y_val * z_val**2) * z_val**2 * (2*t_val)
    term3 = np.sin(x_val) * np.exp(y_val * z_val**2) * (2*y_val*z_val) * (-1/t_val**2)
    term2_vals.append(term2)
    term3_vals.append(term3)

axes[1,0].plot(t_vals, term2_vals, 'r--', linewidth=2, alpha=0.7, label='Term 2: +sin(x)·e·z²·(2t)')
axes[1,0].plot(t_vals, term3_vals, 'g--', linewidth=2, alpha=0.7, label='Term 3: -sin(x)·e·(2yz)·(-1/t²)')
axes[1,0].set_xlabel('t')
axes[1,0].set_ylabel('Term Value')
axes[1,0].set_title('Individual Terms in Total Derivative', fontweight='bold')
axes[1,0].grid(True, alpha=0.3)
axes[1,0].legend()

# Plot 4: Error between approaches (should be zero)
error = np.abs(df_dt_direct_vals - df_dt_total_vals)
axes[1,1].semilogy(t_vals, error + 1e-16, 'purple', linewidth=2)  # +1e-16 to avoid log(0)
axes[1,1].set_xlabel('t')
axes[1,1].set_ylabel('|Error| (log scale)')
axes[1,1].set_title('Error: |Direct - Total Derivative|', fontweight='bold')
axes[1,1].grid(True, alpha=0.3)
axes[1,1].text(0.5, 0.8, f'Max error: {np.max(error):.2e}', 
               transform=axes[1,1].transAxes, fontsize=12,
               bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))

plt.tight_layout()
plt.show()

print("🔍 VISUALIZATION INSIGHTS:")
print("• Top-left: The function oscillates and grows exponentially")
print("• Top-right: Both derivative methods give IDENTICAL results")
print("• Bottom-left: Terms 2 and 3 are mirror images - they cancel perfectly!")
print("• Bottom-right: The error between methods is essentially machine precision")
print()
print("🎯 This demonstrates the power and accuracy of the total derivative method!")
```

```python
# Solutions to Practice Exercises
print("📝 SOLUTIONS TO PRACTICE EXERCISES")
print("=" * 50)

# Define symbols for the new function
u, v, w, t = sp.symbols('u v w t', real=True)
g = u**2 * sp.cos(v*w) + sp.exp(u/v)

print("Function: g(u,v,w) = u²cos(vw) + e^(u/v)")
print()

print("EXERCISE 1: Partial Derivatives")
print("-" * 35)

# Partial derivative with respect to u
dg_du = sp.diff(g, u)
print(f"∂g/∂u = {dg_du}")
print("Explanation: d/du[u²cos(vw)] + d/du[e^(u/v)]")
print("           = 2u·cos(vw) + e^(u/v)·(1/v)")
print()

# Partial derivative with respect to v
dg_dv = sp.diff(g, v)
print(f"∂g/∂v = {dg_dv}")
print("Explanation: d/dv[u²cos(vw)] + d/dv[e^(u/v)]")
print("           = u²·(-sin(vw))·w + e^(u/v)·(-u/v²)")
print()

# Partial derivative with respect to w
dg_dw = sp.diff(g, w)
print(f"∂g/∂w = {dg_dw}")
print("Explanation: d/dw[u²cos(vw)] + d/dw[e^(u/v)]")
print("           = u²·(-sin(vw))·v + 0")
print()

print("EXERCISE 2: Total Derivative")
print("-" * 30)

# Define parametric functions
u_t = 2*t
v_t = t + 1
w_t = t**3

print(f"Given: u(t) = {u_t}, v(t) = {v_t}, w(t) = {w_t}")
print()

# Find derivatives of parametric functions
du_dt = sp.diff(u_t, t)
dv_dt = sp.diff(v_t, t)
dw_dt = sp.diff(w_t, t)

print("Step 1: Find parametric derivatives")
print(f"du/dt = {du_dt}")
print(f"dv/dt = {dv_dt}")
print(f"dw/dt = {dw_dt}")
print()

# Calculate total derivative
print("Step 2: Apply total derivative formula")
print("dg/dt = (∂g/∂u)(du/dt) + (∂g/∂v)(dv/dt) + (∂g/∂w)(dw/dt)")
print()

total_derivative_g = dg_du * du_dt + dg_dv * dv_dt + dg_dw * dw_dt
print(f"dg/dt = ({dg_du})·({du_dt}) + ({dg_dv})·({dv_dt}) + ({dg_dw})·({dw_dt})")
print()

# Simplify
total_derivative_g_simplified = sp.simplify(total_derivative_g)
print(f"Simplified: dg/dt = {total_derivative_g_simplified}")
print()

# Substitute parametric functions
total_derivative_g_sub = total_derivative_g.subs([(u, u_t), (v, v_t), (w, w_t)])
total_derivative_g_final = sp.simplify(total_derivative_g_sub)

print("Step 3: Substitute u(t), v(t), w(t)")
print(f"Final result: dg/dt = {total_derivative_g_final}")
print()

print("🔍 VERIFICATION:")
print("-" * 15)
# Direct substitution for verification
g_direct = g.subs([(u, u_t), (v, v_t), (w, w_t)])
dg_dt_direct = sp.diff(g_direct, t)
print(f"Direct approach: dg/dt = {sp.simplify(dg_dt_direct)}")
print(f"Total derivative:dg/dt = {total_derivative_g_final}")
print()

# Check if equal
verification = sp.simplify(dg_dt_direct - total_derivative_g_final)
if verification == 0:
    print("✅ Perfect match! Both methods give identical results.")
else:
    print(f"Difference: {verification}")
    print("The methods should give the same result - let's double-check the algebra.")

print()
print("🎯 KEY INSIGHTS:")
print("• Partial derivatives of complex functions follow the same rules")
print("• Chain rule applies when exponentials have variable bases and exponents")
print("• Total derivative systematically handles parametric dependencies")
print("• Both direct and total derivative methods must give identical results")
```
