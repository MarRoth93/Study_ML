# The Multivariate Chain Rule

## Introduction

In previous modules, we learned about the **Total Derivative** for multi-variable functions. Now we'll generalize this concept and express it in a more elegant, compact form.

### Recall: The Total Derivative

When we have a function $f(x, y, z)$ where each variable is itself a function of another variable $t$:
- $x = x(t)$
- $y = y(t)$
- $z = z(t)$

The **total derivative** of $f$ with respect to $t$ is:

$$\frac{df}{dt} = \frac{\partial f}{\partial x}\frac{dx}{dt} + \frac{\partial f}{\partial y}\frac{dy}{dt} + \frac{\partial f}{\partial z}\frac{dz}{dt}$$

This is the sum of chains relating $f$ to $t$ through each variable. This piecewise approach is computationally efficient - computers excel at solving problems piece by piece rather than substituting everything at once!

### Our Goal

We'll now:
1. **Generalize** this to functions with any number of variables
2. **Simplify** the notation using vectors and dot products
3. See how this elegant form connects to linear algebra

## Generalizing to n Variables

### The Setup

Consider a function $f$ with $n$ variables: $f(x_1, x_2, x_3, \ldots, x_n)$

Instead of writing out all variables explicitly, we use **vector notation**:

$$f(\mathbf{x})$$

where $\mathbf{x}$ (in bold) represents an $n$-dimensional vector:

$$\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \\ \vdots \\ x_n \end{bmatrix}$$

### The Key Assumption

Each component of $\mathbf{x}$ is itself a function of some variable $t$:
- $x_1 = x_1(t)$
- $x_2 = x_2(t)$
- $x_3 = x_3(t)$
- $\vdots$
- $x_n = x_n(t)$

### The Question

**What is $\frac{df}{dt}$?**

We need to find how $f$ changes with respect to $t$ through all the intermediate variables in $\mathbf{x}$.

## Building the Chain Rule Expression

### Step 1: Gather the Partial Derivatives of f

We need all the partial derivatives of $f$ with respect to each component of $\mathbf{x}$:

$$\frac{\partial f}{\partial x_1}, \frac{\partial f}{\partial x_2}, \frac{\partial f}{\partial x_3}, \ldots, \frac{\partial f}{\partial x_n}$$

Let's organize these into a **vector** (this is the gradient or Jacobian):

$$\nabla f = \begin{bmatrix} \frac{\partial f}{\partial x_1} \\ \frac{\partial f}{\partial x_2} \\ \frac{\partial f}{\partial x_3} \\ \vdots \\ \frac{\partial f}{\partial x_n} \end{bmatrix}$$

### Step 2: Gather the Derivatives of x with respect to t

We need the rate of change of each component of $\mathbf{x}$ with respect to $t$:

$$\frac{dx_1}{dt}, \frac{dx_2}{dt}, \frac{dx_3}{dt}, \ldots, \frac{dx_n}{dt}$$

Let's organize these into another **vector**:

$$\frac{d\mathbf{x}}{dt} = \begin{bmatrix} \frac{dx_1}{dt} \\ \frac{dx_2}{dt} \\ \frac{dx_3}{dt} \\ \vdots \\ \frac{dx_n}{dt} \end{bmatrix}$$

### Step 3: Combine Them Using the Chain Rule

We want the **sum of products** of corresponding terms from each vector:

$$\frac{df}{dt} = \frac{\partial f}{\partial x_1}\frac{dx_1}{dt} + \frac{\partial f}{\partial x_2}\frac{dx_2}{dt} + \cdots + \frac{\partial f}{\partial x_n}\frac{dx_n}{dt}$$

**Wait!** This is exactly what the **dot product** does! 🎯

## The Elegant Final Form ✨

### The Multivariate Chain Rule

$$\boxed{\frac{df}{dt} = \nabla f \cdot \frac{d\mathbf{x}}{dt}}$$

or equivalently:

$$\boxed{\frac{df}{dt} = \frac{\partial f}{\partial \mathbf{x}} \cdot \frac{d\mathbf{x}}{dt}}$$

where $\cdot$ represents the **dot product** of the two vectors.

### Why This is Beautiful

1. **Compact notation**: No need to write out all $n$ terms explicitly
2. **Connects to linear algebra**: Uses the dot product we already know
3. **Easy to remember**: Gradient of $f$ "dot" rate of change of $\mathbf{x}$
4. **Computationally efficient**: Perfect for programming

### Visual Intuition

Think of it this way:
- $\nabla f$ tells us **how $f$ changes** in each direction in $\mathbf{x}$-space
- $\frac{d\mathbf{x}}{dt}$ tells us **how $\mathbf{x}$ is moving** as $t$ changes
- The dot product combines these to give the **total rate of change** of $f$ with respect to $t$

## Example 1: Simple 2D Case

Let's start with a concrete example to see how this works.

### Problem Setup

Given:
- $f(x, y) = x^2 + 3xy + y^2$
- $x(t) = \cos(t)$
- $y(t) = \sin(t)$

**Find:** $\frac{df}{dt}$

### Method 1: The Long Way (Substitution)

We could substitute $x(t)$ and $y(t)$ into $f$, then differentiate. Let's see how tedious that would be!

````python
# Define symbolic variables
t = sp.Symbol('t')
x, y = sp.symbols('x y')

# Define the function and parametric equations
f = x**2 + 3*x*y + y**2
x_t = sp.cos(t)
y_t = sp.sin(t)

print("="*60)
print("METHOD 1: Direct Substitution (The Hard Way)")
print("="*60)

# Substitute x(t) and y(t) into f
f_of_t = f.subs({x: x_t, y: y_t})
print(f"\nf(t) = f(cos(t), sin(t)) = {f_of_t}")

# Now differentiate
df_dt_method1 = sp.diff(f_of_t, t)
print(f"\ndf/dt = {df_dt_method1}")

# Simplify
df_dt_method1_simplified = sp.simplify(df_dt_method1)
print(f"\nSimplified: df/dt = {df_dt_method1_simplified}")
print("\n⚠️  This substitution method gets messy very quickly!")
````

### Method 2: Using the Multivariate Chain Rule (The Smart Way)

Using our elegant formula: $\frac{df}{dt} = \nabla f \cdot \frac{d\mathbf{x}}{dt}$

**Step 1:** Compute $\nabla f = \begin{bmatrix} \frac{\partial f}{\partial x} \\ \frac{\partial f}{\partial y} \end{bmatrix}$

**Step 2:** Compute $\frac{d\mathbf{x}}{dt} = \begin{bmatrix} \frac{dx}{dt} \\ \frac{dy}{dt} \end{bmatrix}$

**Step 3:** Take the dot product and substitute values

````python
print("="*60)
print("METHOD 2: Multivariate Chain Rule (The Smart Way)")
print("="*60)

# Step 1: Compute the gradient of f
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

print("\nStep 1: Compute ∇f")
print(f"∂f/∂x = {df_dx}")
print(f"∂f/∂y = {df_dy}")

gradient_f = sp.Matrix([df_dx, df_dy])
print(f"\n∇f = {gradient_f.T}")

# Step 2: Compute dx/dt and dy/dt
dx_dt = sp.diff(x_t, t)
dy_dt = sp.diff(y_t, t)

print("\nStep 2: Compute dx/dt")
print(f"dx/dt = {dx_dt}")
print(f"dy/dt = {dy_dt}")

dx_vector = sp.Matrix([dx_dt, dy_dt])
print(f"\ndx/dt = {dx_vector.T}")

# Step 3: Compute the dot product
print("\nStep 3: Compute the dot product ∇f · (dx/dt)")
df_dt_method2 = gradient_f.dot(dx_vector)
print(f"\ndf/dt = ({df_dx}) · ({dx_dt}) + ({df_dy}) · ({dy_dt})")
print(f"      = {df_dt_method2}")

# Now substitute x(t) and y(t)
df_dt_method2_sub = df_dt_method2.subs({x: x_t, y: y_t})
print(f"\nSubstituting x = cos(t), y = sin(t):")
print(f"df/dt = {df_dt_method2_sub}")

# Simplify
df_dt_method2_simplified = sp.simplify(df_dt_method2_sub)
print(f"\nSimplified: df/dt = {df_dt_method2_simplified}")

print("\n" + "="*60)
print("VERIFICATION: Both methods give the same result!")
print("="*60)
print(f"Method 1: {df_dt_method1_simplified}")
print(f"Method 2: {df_dt_method2_simplified}")
print(f"Are they equal? {sp.simplify(df_dt_method1_simplified - df_dt_method2_simplified) == 0} ✓")
````

## Example 2: Three-Variable Function

Let's work with a function of three variables to see the full power of vector notation.

### Problem Setup

Given:
- $f(x, y, z) = x^2yz + z^3$
- $x(t) = t^2$
- $y(t) = 2t$
- $z(t) = t^3$

**Find:** $\frac{df}{dt}$ using the multivariate chain rule

### Solution

Using: $\frac{df}{dt} = \nabla f \cdot \frac{d\mathbf{x}}{dt}$

where $\mathbf{x} = \begin{bmatrix} x \\ y \\ z \end{bmatrix}$ and $\nabla f = \begin{bmatrix} \frac{\partial f}{\partial x} \\ \frac{\partial f}{\partial y} \\ \frac{\partial f}{\partial z} \end{bmatrix}$

````python
# Define new symbols for 3D case
t = sp.Symbol('t')
x, y, z = sp.symbols('x y z')

# Define the function
f_3d = x**2 * y * z + z**3

# Define parametric functions
x_t = t**2
y_t = 2*t
z_t = t**3

print("="*70)
print("THREE-VARIABLE CHAIN RULE EXAMPLE")
print("="*70)

print(f"\nFunction: f(x, y, z) = {f_3d}")
print(f"x(t) = {x_t}")
print(f"y(t) = {y_t}")
print(f"z(t) = {z_t}")

print("\n" + "="*70)
print("STEP 1: Compute the gradient ∇f")
print("="*70)

df_dx = sp.diff(f_3d, x)
df_dy = sp.diff(f_3d, y)
df_dz = sp.diff(f_3d, z)

print(f"∂f/∂x = {df_dx}")
print(f"∂f/∂y = {df_dy}")
print(f"∂f/∂z = {df_dz}")

grad_f = sp.Matrix([df_dx, df_dy, df_dz])
print(f"\n∇f = {grad_f.T}")

print("\n" + "="*70)
print("STEP 2: Compute dx/dt")
print("="*70)

dx_dt = sp.diff(x_t, t)
dy_dt = sp.diff(y_t, t)
dz_dt = sp.diff(z_t, t)

print(f"dx/dt = {dx_dt}")
print(f"dy/dt = {dy_dt}")
print(f"dz/dt = {dz_dt}")

dx_dt_vector = sp.Matrix([dx_dt, dy_dt, dz_dt])
print(f"\ndx/dt = {dx_dt_vector.T}")

print("\n" + "="*70)
print("STEP 3: Compute the dot product")
print("="*70)

df_dt = grad_f.dot(dx_dt_vector)
print(f"\ndf/dt = ∇f · (dx/dt)")
print(f"      = ({df_dx})·({dx_dt}) + ({df_dy})·({dy_dt}) + ({df_dz})·({dz_dt})")
print(f"      = {df_dt}")

print("\n" + "="*70)
print("STEP 4: Substitute x(t), y(t), z(t)")
print("="*70)

df_dt_substituted = df_dt.subs({x: x_t, y: y_t, z: z_t})
print(f"\ndf/dt = {df_dt_substituted}")

# Expand and simplify
df_dt_expanded = sp.expand(df_dt_substituted)
print(f"\nExpanded: df/dt = {df_dt_expanded}")

# Factor if possible
df_dt_factored = sp.factor(df_dt_expanded)
print(f"Factored: df/dt = {df_dt_factored}")

print("\n✓ Done! The chain rule made this calculation straightforward.")
````

## The Multivariate Chain Rule Works with All Our Other Rules! 🎉

An important insight from the transcript: **All the time-saving differentiation rules we learned still work for multivariate problems!**

### Our Toolkit Still Applies

The following rules work seamlessly with the multivariate chain rule:

1. **Product Rule**: $\frac{d}{dt}[f(\mathbf{x}) \cdot g(\mathbf{x})] = f'(\mathbf{x}) \cdot g(\mathbf{x}) + f(\mathbf{x}) \cdot g'(\mathbf{x})$

2. **Quotient Rule**: $\frac{d}{dt}\left[\frac{f(\mathbf{x})}{g(\mathbf{x})}\right] = \frac{f'(\mathbf{x}) \cdot g(\mathbf{x}) - f(\mathbf{x}) \cdot g'(\mathbf{x})}{[g(\mathbf{x})]^2}$

3. **Power Rule**: Still applies to each component

4. **Sum/Difference Rules**: Derivatives distribute over addition and subtraction

### Why This Matters

You don't need to learn new rules! The chain rule just gives us a systematic way to handle the dependencies between variables. Once you have that, all your familiar differentiation tools still work.

## Visual Summary: The Chain of Dependencies

### The Dependency Chain

```
      t
      ↓
   ┌──┴──┐
   ↓     ↓     ↓
  x₁    x₂   ... xₙ
   ↓     ↓     ↓
   └──┬──┘
      ↓
      f
```

The chain rule tells us how changes in $t$ propagate through all the intermediate variables to affect $f$.

### The Formula (Multiple Notations)

All of these mean the same thing:

$$\frac{df}{dt} = \nabla f \cdot \frac{d\mathbf{x}}{dt}$$

$$\frac{df}{dt} = \frac{\partial f}{\partial \mathbf{x}} \cdot \frac{d\mathbf{x}}{dt}$$

$$\frac{df}{dt} = \sum_{i=1}^{n} \frac{\partial f}{\partial x_i} \frac{dx_i}{dt}$$

$$\frac{df}{dt} = \frac{\partial f}{\partial x_1}\frac{dx_1}{dt} + \frac{\partial f}{\partial x_2}\frac{dx_2}{dt} + \cdots + \frac{\partial f}{\partial x_n}\frac{dx_n}{dt}$$

### Key Insight

The **dot product** notation is most elegant because:
- It's compact
- It leverages linear algebra we already know
- It's easy to implement computationally
- It generalizes naturally to any number of variables

## Why Computers Love the Chain Rule

The transcript mentions that **computers are really good at solving piecewise problems quickly**. Let's see why the chain rule is computationally efficient.

````python
# Let's implement both approaches numerically and compare
import time

def method1_substitution(t_val):
    """Direct substitution then differentiation (conceptually)"""
    # This would involve substituting first, then taking derivative
    # For numerical evaluation, we'd need to do more work
    x = np.cos(t_val)
    y = np.sin(t_val)
    # We'd need finite differences or symbolic computation
    return None  # Placeholder

def method2_chain_rule(t_val):
    """Chain rule approach - compute pieces separately"""
    # Compute x(t) and y(t)
    x = np.cos(t_val)
    y = np.sin(t_val)
    
    # Compute partial derivatives of f at (x,y)
    df_dx = 2*x + 3*y
    df_dy = 3*x + 2*y
    
    # Compute dx/dt and dy/dt
    dx_dt = -np.sin(t_val)
    dy_dt = np.cos(t_val)
    
    # Dot product
    df_dt = df_dx * dx_dt + df_dy * dy_dt
    
    return df_dt

# Test at several points
t_test = np.array([0, np.pi/4, np.pi/2, np.pi])

print("="*70)
print("COMPUTATIONAL EFFICIENCY DEMONSTRATION")
print("="*70)

print("\nChain Rule Approach (modular and efficient):")
print("-" * 70)

for t_val in t_test:
    result = method2_chain_rule(t_val)
    print(f"t = {t_val:6.4f} → df/dt = {result:8.4f}")

print("\n" + "="*70)
print("WHY THE CHAIN RULE IS BETTER FOR COMPUTATION:")
print("="*70)
print("""
1. MODULAR: Break complex problem into simple pieces
   - Compute partial derivatives of f
   - Compute derivatives of x(t), y(t), etc.
   - Combine with dot product
   
2. REUSABLE: Can compute ∇f once and use it for different paths
   
3. PARALLELIZABLE: Each partial derivative can be computed independently
   
4. NUMERICALLY STABLE: Avoid complex symbolic manipulations
   
5. MEMORY EFFICIENT: Don't need to store entire substituted expression

This is why automatic differentiation (used in deep learning) 
is based on the chain rule!
""")
````

## Key Takeaways

### The Multivariate Chain Rule Formula

$$\boxed{\frac{df}{dt} = \nabla f \cdot \frac{d\mathbf{x}}{dt} = \sum_{i=1}^{n} \frac{\partial f}{\partial x_i} \frac{dx_i}{dt}}$$

### When to Use It

Use the multivariate chain rule when:
- You have a function $f$ that depends on multiple variables
- Those variables themselves depend on another variable (often time $t$)
- You want to find how $f$ changes with respect to that other variable

### The Three-Step Algorithm

1. **Compute the gradient** $\nabla f$ (all partial derivatives of $f$)
2. **Compute the velocity vector** $\frac{d\mathbf{x}}{dt}$ (derivatives of each variable)
3. **Take the dot product** to get $\frac{df}{dt}$

### Why This Matters

1. **Efficiency**: Breaks complex derivatives into manageable pieces
2. **Modularity**: Each piece can be computed independently
3. **Generalization**: Works for any number of variables
4. **Computation**: Perfect for automatic differentiation in machine learning
5. **Clarity**: Clean notation makes the math easier to follow

### Connection to Machine Learning

The multivariate chain rule is the **foundation of backpropagation** in neural networks:
- Neural networks are compositions of many functions
- Backpropagation uses the chain rule to compute gradients efficiently
- This enables gradient descent to train deep networks

### Remember

> **All your familiar differentiation rules still work!**
>
> The multivariate chain rule doesn't replace other rules—it complements them. You can still use:
> - Product rule
> - Quotient rule  
> - Power rule
> - Etc.
>
> The chain rule just handles the dependencies between variables.

## Summary: The Big Picture

### From Total Derivative to General Form

We started with the **total derivative** for specific cases like $f(x, y, z)$:

$$\frac{df}{dt} = \frac{\partial f}{\partial x}\frac{dx}{dt} + \frac{\partial f}{\partial y}\frac{dy}{dt} + \frac{\partial f}{\partial z}\frac{dz}{dt}$$

We generalized it to **any number of variables** using vector notation:

$$\frac{df}{dt} = \nabla f \cdot \frac{d\mathbf{x}}{dt}$$

### The Power of Abstraction

By using **bold notation** for vectors:
- $\mathbf{x}$ represents the entire vector $(x_1, x_2, \ldots, x_n)$
- $\nabla f$ represents all partial derivatives
- The dot product captures the sum of all the chain terms

This abstraction:
- **Simplifies notation** (no need to write out all $n$ terms)
- **Clarifies the concept** (it's about how changes propagate)
- **Enables computation** (perfect for programming)
- **Connects to linear algebra** (uses familiar operations)
