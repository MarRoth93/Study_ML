```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sympy import *
import warnings
warnings.filterwarnings('ignore')

# Configure matplotlib for better plots
plt.style.use('default')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# Define symbolic variables
x = symbols('x')

print("✅ Libraries imported successfully!")
print("Ready to tackle complex derivatives!")
```

## Step 1: Rewrite as a Product 🔄

The first strategic move is to rewrite our fraction as a product. This makes it easier to apply our differentiation rules.

### Original Function:
$$f(x) = \frac{\sin(2x^5 + 3x)}{e^{7x}}$$

### Rewritten as Product:
$$f(x) = \sin(2x^5 + 3x) \cdot e^{-7x}$$

### Why This Helps:
- **Product Rule**: We can use the product rule instead of the more complex quotient rule
- **Simpler Pattern**: $\frac{1}{e^{7x}} = e^{-7x}$ is easier to differentiate
- **Fewer Rules to Remember**: We avoid memorizing the quotient rule formula

### The Mathematical Transformation:
$$\frac{A}{B} = A \cdot B^{-1}$$

So: $\frac{\sin(2x^5 + 3x)}{e^{7x}} = \sin(2x^5 + 3x) \cdot (e^{7x})^{-1} = \sin(2x^5 + 3x) \cdot e^{-7x}$

```python
# Define our original function and its rewritten form
original_function = sin(2*x**5 + 3*x) / exp(7*x)
rewritten_function = sin(2*x**5 + 3*x) * exp(-7*x)

print("Original function:")
print(f"f(x) = {original_function}")
print()
print("Rewritten as product:")
print(f"f(x) = {rewritten_function}")
print()

# Verify they're the same by simplifying
difference = simplify(original_function - rewritten_function)
print(f"Difference between forms: {difference}")
print()
print("✅ The two forms are mathematically equivalent!")

# Let's identify the two parts for the product rule
g_x = sin(2*x**5 + 3*x)  # First part
h_x = exp(-7*x)          # Second part

print()
print("For the product rule, we have:")
print(f"g(x) = {g_x}")
print(f"h(x) = {h_x}")
print()
print("Now we need to find g'(x) and h'(x) separately...")
```

## Step 2: Differentiate the First Part - g(x) 🎯

Now we need to find the derivative of the first part:
$$g(x) = \sin(2x^5 + 3x)$$

### Recognizing the Pattern
This is a **composite function** - perfect for the **chain rule**!

- **Outer function**: $\sin(u)$ where $u = 2x^5 + 3x$
- **Inner function**: $u(x) = 2x^5 + 3x$

### Chain Rule Application
For $g(x) = \sin(u(x))$:
$$g'(x) = \frac{d}{dx}[\sin(u)] = \cos(u) \cdot \frac{du}{dx}$$

### Step-by-Step Calculation

1. **Outer derivative**: $\frac{d}{du}[\sin(u)] = \cos(u)$

2. **Inner derivative**: $\frac{du}{dx} = \frac{d}{dx}[2x^5 + 3x]$
   - Using **sum rule**: $\frac{d}{dx}[2x^5] + \frac{d}{dx}[3x]$
   - Using **power rule**: $2 \cdot 5x^4 + 3 = 10x^4 + 3$

3. **Combine using chain rule**:
   $$g'(x) = \cos(2x^5 + 3x) \cdot (10x^4 + 3)$$

### Rules Used So Far:
- ✅ **Chain Rule** (main structure)
- ✅ **Sum Rule** (for the polynomial inside)
- ✅ **Power Rule** (for individual terms)

```python
# Step-by-step calculation of g'(x)
print("🔍 CALCULATING g'(x) = d/dx[sin(2x⁵ + 3x)]")
print("="*50)

# Define the components
g_x = sin(2*x**5 + 3*x)
u = 2*x**5 + 3*x  # Inner function

print("1. Identify the structure:")
print(f"   g(x) = sin(u) where u = {u}")
print()

print("2. Apply chain rule: g'(x) = d/du[sin(u)] × du/dx")
print()

# Outer derivative
outer_deriv = cos(u)
print(f"3. Outer derivative: d/du[sin(u)] = cos(u) = cos({u})")
print()

# Inner derivative
inner_deriv = diff(u, x)
print(f"4. Inner derivative: du/dx = d/dx[{u}]")
print(f"   Using sum rule: d/dx[2x⁵] + d/dx[3x]")
print(f"   Using power rule: 2×5x⁴ + 3 = {inner_deriv}")
print()

# Combine using chain rule
g_prime = diff(g_x, x)
print(f"5. Chain rule result:")
print(f"   g'(x) = cos({u}) × ({inner_deriv})")
print(f"   g'(x) = {g_prime}")
print()

# Let's also verify with SymPy
sympy_result = diff(g_x, x)
print(f"✅ SymPy verification: {sympy_result}")
print()
print("🎉 First part complete! We have g'(x).")
```

## Step 3: Differentiate the Second Part - h(x) ⚡

Now for the second part of our product:
$$h(x) = e^{-7x}$$

### Recognizing the Pattern
This is also a **composite function** - another application of the **chain rule**!

- **Outer function**: $e^v$ where $v = -7x$
- **Inner function**: $v(x) = -7x$

### Chain Rule Application
For $h(x) = e^{v(x)}$:
$$h'(x) = \frac{d}{dx}[e^v] = e^v \cdot \frac{dv}{dx}$$

### Step-by-Step Calculation

1. **Outer derivative**: $\frac{d}{dv}[e^v] = e^v$

2. **Inner derivative**: $\frac{dv}{dx} = \frac{d}{dx}[-7x] = -7$

3. **Combine using chain rule**:
   $$h'(x) = e^{-7x} \cdot (-7) = -7e^{-7x}$$

### Key Insight
Notice how the exponential function $e^{something}$ always differentiates to itself times the derivative of the "something". This makes exponentials particularly nice to work with!

### Rules Used:
- ✅ **Chain Rule** (main structure)  
- ✅ **Power Rule** (for the linear function $-7x$)

```python
# Step-by-step calculation of h'(x)
print("🔍 CALCULATING h'(x) = d/dx[e^(-7x)]")
print("="*40)

# Define the components
h_x = exp(-7*x)
v = -7*x  # Inner function

print("1. Identify the structure:")
print(f"   h(x) = e^v where v = {v}")
print()

print("2. Apply chain rule: h'(x) = d/dv[e^v] × dv/dx")
print()

# Outer derivative
print(f"3. Outer derivative: d/dv[e^v] = e^v = e^({v})")
print()

# Inner derivative
inner_deriv_h = diff(v, x)
print(f"4. Inner derivative: dv/dx = d/dx[{v}] = {inner_deriv_h}")
print()

# Combine using chain rule
h_prime = diff(h_x, x)
print(f"5. Chain rule result:")
print(f"   h'(x) = e^({v}) × ({inner_deriv_h})")
print(f"   h'(x) = {h_prime}")
print()

# Verify with SymPy
sympy_h_result = diff(h_x, x)
print(f"✅ SymPy verification: {sympy_h_result}")
print()
print("🎉 Second part complete! We have h'(x).")
print()
print("📋 SUMMARY SO FAR:")
print("-" * 30)
print(f"g(x) = {g_x}")
print(f"g'(x) = {diff(g_x, x)}")
print()
print(f"h(x) = {h_x}")  
print(f"h'(x) = {h_prime}")
print()
print("Next: Apply the product rule to combine everything!")
```

## Step 4: Apply the Product Rule 🚀

Now we combine our results using the **product rule**!

### Product Rule Formula
For $f(x) = g(x) \cdot h(x)$:
$$f'(x) = g'(x) \cdot h(x) + g(x) \cdot h'(x)$$

### Our Components
- $g(x) = \sin(2x^5 + 3x)$
- $g'(x) = \cos(2x^5 + 3x) \cdot (10x^4 + 3)$
- $h(x) = e^{-7x}$
- $h'(x) = -7e^{-7x}$

### Applying the Product Rule
$$f'(x) = g'(x) \cdot h(x) + g(x) \cdot h'(x)$$

$$f'(x) = [\cos(2x^5 + 3x) \cdot (10x^4 + 3)] \cdot e^{-7x} + \sin(2x^5 + 3x) \cdot (-7e^{-7x})$$

### Simplifying
$$f'(x) = (10x^4 + 3)\cos(2x^5 + 3x)e^{-7x} - 7\sin(2x^5 + 3x)e^{-7x}$$

We can factor out $e^{-7x}$:
$$f'(x) = e^{-7x}[(10x^4 + 3)\cos(2x^5 + 3x) - 7\sin(2x^5 + 3x)]$$

### All Four Rules Used! 🏆
- ✅ **Sum Rule** (inside the polynomial $2x^5 + 3x$)
- ✅ **Power Rule** (for differentiating $x^5$ and $x$)
- ✅ **Chain Rule** (for both composite functions)
- ✅ **Product Rule** (to combine the two parts)

```python
# Final calculation using the product rule
print("🎯 FINAL CALCULATION: Product Rule")
print("="*50)

# Our function components
f_x = sin(2*x**5 + 3*x) * exp(-7*x)
g_x = sin(2*x**5 + 3*x)
h_x = exp(-7*x)

# Their derivatives (calculated earlier)
g_prime = diff(g_x, x)
h_prime = diff(h_x, x)

print("Product Rule: f'(x) = g'(x)×h(x) + g(x)×h'(x)")
print()
print("Components:")
print(f"g(x) = {g_x}")
print(f"g'(x) = {g_prime}")
print(f"h(x) = {h_x}")
print(f"h'(x) = {h_prime}")
print()

# Apply product rule step by step
term1 = g_prime * h_x
term2 = g_x * h_prime

print("Step-by-step product rule application:")
print()
print("Term 1: g'(x) × h(x)")
print(f"= {g_prime} × {h_x}")
print(f"= {term1}")
print()

print("Term 2: g(x) × h'(x)")
print(f"= {g_x} × {h_prime}")
print(f"= {term2}")
print()

# Final result
f_prime = term1 + term2
print("Final result: f'(x) = Term 1 + Term 2")
print(f"f'(x) = {f_prime}")
print()

# Let's factor out common terms for a cleaner look
f_prime_factored = factor(f_prime)
print("Factored form:")
print(f"f'(x) = {f_prime_factored}")
print()

# Verify with direct differentiation
direct_result = diff(f_x, x)
print("✅ Verification by direct differentiation:")
print(f"f'(x) = {direct_result}")
print()

# Check if they're the same
difference = simplify(f_prime - direct_result)
print(f"Difference between methods: {difference}")
print()
print("🎉 SUCCESS! Our step-by-step approach matches direct differentiation!")
```

## Visualization: Seeing the Mathematics 📊

Let's visualize our complex function and its derivative to gain intuition about their behavior.

```python
# Create visualizations of our function and its derivative
print("📊 Creating visualizations...")

# Convert symbolic expressions to numerical functions
from sympy import lambdify

# Our functions
f_original = sin(2*x**5 + 3*x) / exp(7*x)  # Original form
f_product = sin(2*x**5 + 3*x) * exp(-7*x)   # Product form
f_derivative = diff(f_product, x)

# Convert to numerical functions
f_func = lambdify(x, f_product, 'numpy')
f_prime_func = lambdify(x, f_derivative, 'numpy')

# Create x values for plotting (be careful with the range due to exponential behavior)
x_vals = np.linspace(-0.5, 0.5, 1000)

# Calculate y values
try:
    y_vals = f_func(x_vals)
    y_prime_vals = f_prime_func(x_vals)
    
    # Create subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))
    
    # Plot the original function
    ax1.plot(x_vals, y_vals, 'b-', linewidth=2, label=r'$f(x) = \frac{\sin(2x^5 + 3x)}{e^{7x}}$')
    ax1.set_title('Original Function: Complex but Manageable!', fontsize=14, fontweight='bold')
    ax1.set_xlabel('x')
    ax1.set_ylabel('f(x)')
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    ax1.set_xlim(-0.5, 0.5)
    
    # Plot the derivative
    ax2.plot(x_vals, y_prime_vals, 'r-', linewidth=2, label="f'(x) (our calculated derivative)")
    ax2.set_title("Derivative: Result of Combining All Four Rules", fontsize=14, fontweight='bold')
    ax2.set_xlabel('x')
    ax2.set_ylabel("f'(x)")
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    ax2.set_xlim(-0.5, 0.5)
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Visualizations created successfully!")
    print()
    print("🔍 OBSERVATIONS:")
    print("• The original function shows complex oscillatory behavior due to the sine")
    print("• The exponential denominator causes rapid decay")
    print("• The derivative captures how this complex function changes")
    print("• Even 'scary' functions have well-defined, calculable derivatives!")
    
except Exception as e:
    print(f"Note: Visualization requires careful handling due to exponential growth.")
    print(f"The mathematical result is still valid: {f_derivative}")
    
# Let's also show a simpler range
print()
print("📈 Function behavior near x = 0:")
test_points = [-0.1, -0.05, 0, 0.05, 0.1]
print("x\t\tf(x)\t\tf'(x)")
print("-" * 40)
for point in test_points:
    f_val = float(f_func(point))
    f_prime_val = float(f_prime_func(point))
    print(f"{point:.2f}\t\t{f_val:.6f}\t{f_prime_val:.6f}")
```
