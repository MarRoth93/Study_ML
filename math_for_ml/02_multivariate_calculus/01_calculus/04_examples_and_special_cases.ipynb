## Special Case 1: The Reciprocal Function $f(x) = \frac{1}{x}$

### Understanding the Function
The function $f(x) = \frac{1}{x}$ has some fascinating properties:

- **Negative gradient everywhere** (except at $x = 0$)
- **Discontinuity at $x = 0$** - the function is undefined there
- **Approaches infinity** as $x$ approaches 0 from either side
- **Never crosses the x-axis** - always positive for $x > 0$, always negative for $x < 0$

### The Mystery at $x = 0$
What happens at $x = 0$? The function "jumps" from $-\infty$ to $+\infty$, creating a **discontinuity**. This is where the function simply doesn't exist - division by zero is undefined!

```python
# Visualize the reciprocal function f(x) = 1/x
import numpy as np
import matplotlib.pyplot as plt

# Create x values, avoiding x = 0
x_positive = np.linspace(0.1, 5, 1000)
x_negative = np.linspace(-5, -0.1, 1000)

# Calculate y = 1/x
y_positive = 1 / x_positive
y_negative = 1 / x_negative

# Create the plot
plt.figure(figsize=(12, 8))

# Plot both branches
plt.plot(x_positive, y_positive, 'b-', linewidth=3, label='f(x) = 1/x (x > 0)')
plt.plot(x_negative, y_negative, 'b-', linewidth=3, label='f(x) = 1/x (x < 0)')

# Add asymptotes
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.7, label='Horizontal asymptote (y = 0)')
plt.axvline(x=0, color='gray', linestyle='--', alpha=0.7, label='Vertical asymptote (x = 0)')

# Highlight the discontinuity
plt.plot(0, 0, 'ro', markersize=10, markerfacecolor='white', markeredgecolor='red', 
         markeredgewidth=3, label='Discontinuity at x = 0')

# Add arrows to show behavior near discontinuity
plt.annotate('Approaches +∞', xy=(0.2, 5), xytext=(1, 8),
            arrowprops=dict(arrowstyle='->', color='red', lw=2),
            fontsize=12, color='red')
plt.annotate('Approaches -∞', xy=(-0.2, -5), xytext=(-1, -8),
            arrowprops=dict(arrowstyle='->', color='red', lw=2),
            fontsize=12, color='red')

plt.xlim(-5, 5)
plt.ylim(-10, 10)
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('The Reciprocal Function: f(x) = 1/x\nNotice the negative gradient everywhere!', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()

# Add text annotations about gradient
plt.text(2, 2, 'Negative slope\n(decreasing)', fontsize=11, 
         bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.7))
plt.text(-2, -2, 'Negative slope\n(decreasing)', fontsize=11,
         bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.7))

plt.tight_layout()
plt.show()

print("Key Observations:")
print("• The function has a negative gradient everywhere it exists")
print("• There's a discontinuity (break) at x = 0")
print("• The function approaches ±∞ as x approaches 0")
print("• This creates two separate 'branches' of the function")
```

### Differentiating $f(x) = \frac{1}{x}$ Using the Formal Definition

Let's apply our formal definition to find the derivative step by step:

$$f'(x) = \lim_{\Delta x \to 0} \frac{f(x + \Delta x) - f(x)}{\Delta x}$$

**Step 1**: Substitute $f(x) = \frac{1}{x}$
$$f'(x) = \lim_{\Delta x \to 0} \frac{\frac{1}{x + \Delta x} - \frac{1}{x}}{\Delta x}$$

**Step 2**: Combine the fractions in the numerator (find common denominator)
$$f'(x) = \lim_{\Delta x \to 0} \frac{\frac{x}{x(x + \Delta x)} - \frac{x + \Delta x}{x(x + \Delta x)}}{\Delta x}$$

**Step 3**: Simplify the numerator
$$f'(x) = \lim_{\Delta x \to 0} \frac{\frac{x - (x + \Delta x)}{x(x + \Delta x)}}{\Delta x}$$

$$f'(x) = \lim_{\Delta x \to 0} \frac{\frac{-\Delta x}{x(x + \Delta x)}}{\Delta x}$$

**Step 4**: Cancel $\Delta x$
$$f'(x) = \lim_{\Delta x \to 0} \frac{-\Delta x}{x(x + \Delta x)} \cdot \frac{1}{\Delta x}$$

$$f'(x) = \lim_{\Delta x \to 0} \frac{-1}{x(x + \Delta x)}$$

**Step 5**: Take the limit (as $\Delta x \to 0$)
$$f'(x) = \frac{-1}{x \cdot x} = -\frac{1}{x^2}$$

### Result: The Derivative
$$\boxed{\frac{d}{dx}\left(\frac{1}{x}\right) = -\frac{1}{x^2}}$$

```python
# Visualize f(x) = 1/x and its derivative f'(x) = -1/x²

# Create x values (avoiding x = 0)
x_pos = np.linspace(0.2, 4, 1000)
x_neg = np.linspace(-4, -0.2, 1000)

# Original function f(x) = 1/x
f_pos = 1 / x_pos
f_neg = 1 / x_neg

# Derivative f'(x) = -1/x²
f_prime_pos = -1 / (x_pos**2)
f_prime_neg = -1 / (x_neg**2)

# Create side-by-side plots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Plot original function
ax1.plot(x_pos, f_pos, 'b-', linewidth=3, label='f(x) = 1/x')
ax1.plot(x_neg, f_neg, 'b-', linewidth=3)
ax1.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
ax1.axvline(x=0, color='gray', linestyle='--', alpha=0.5)
ax1.set_xlim(-4, 4)
ax1.set_ylim(-5, 5)
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')
ax1.set_title('Original Function: f(x) = 1/x')
ax1.grid(True, alpha=0.3)
ax1.legend()

# Plot derivative
ax2.plot(x_pos, f_prime_pos, 'r-', linewidth=3, label="f'(x) = -1/x²")
ax2.plot(x_neg, f_prime_neg, 'r-', linewidth=3)
ax2.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
ax2.axvline(x=0, color='gray', linestyle='--', alpha=0.5)
ax2.set_xlim(-4, 4)
ax2.set_ylim(-5, 0.5)
ax2.set_xlabel('x')
ax2.set_ylabel("f'(x)")
ax2.set_title("Derivative: f'(x) = -1/x²")
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

# Verify our derivative using numerical approximation
def numerical_derivative(f, x, h=1e-8):
    return (f(x + h) - f(x)) / h

def f(x):
    return 1 / x

def f_prime_analytical(x):
    return -1 / (x**2)

print("Verification: Analytical vs Numerical Derivative")
print("="*50)
test_points = [0.5, 1, 2, -0.5, -1, -2]

for x in test_points:
    analytical = f_prime_analytical(x)
    numerical = numerical_derivative(f, x)
    error = abs(analytical - numerical)
    print(f"x = {x:4.1f} | Analytical: {analytical:8.4f} | Numerical: {numerical:8.4f} | Error: {error:.2e}")

print("\nKey Properties of the Derivative:")
print("• f'(x) = -1/x² is ALWAYS negative (confirms negative gradient)")
print("• f'(x) is undefined at x = 0 (same as original function)")
print("• f'(x) approaches -∞ as x approaches 0 (steeper and steeper slope)")
print("• The derivative is also discontinuous at x = 0")
```

## Special Case 2: The Exponential Function $f(x) = e^x$

The exponential function $e^x$ is perhaps the most remarkable function in mathematics when it comes to differentiation. It has a **unique self-similarity property**: it is its own derivative!

### Key Property
$$\frac{d}{dx}[e^x] = e^x$$

This means that:
- The **rate of change** of $e^x$ at any point equals the **value** of the function at that point
- If $f(x) = e^x$ and $f'(x) = e^x$, then the slope of the tangent line equals the y-coordinate
- This self-replication property makes $e^x$ fundamental in modeling exponential growth

### Why is this special?

1. **Growth Rate = Current Value**: The exponential function grows at a rate proportional to its current size
2. **Perfect Self-Similarity**: No other elementary function has this property
3. **Natural Base**: This is why $e ≈ 2.718...$ is called the "natural" base

### Mathematical Proof (using the limit definition)

Starting with the formal definition:
$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

For $f(x) = e^x$:
$$f'(x) = \lim_{h \to 0} \frac{e^{x+h} - e^x}{h}$$

Factor out $e^x$:
$$f'(x) = \lim_{h \to 0} \frac{e^x \cdot e^h - e^x}{h} = e^x \lim_{h \to 0} \frac{e^h - 1}{h}$$

The key insight is that $\lim_{h \to 0} \frac{e^h - 1}{h} = 1$ (this is actually the **definition** of $e$!):

$$f'(x) = e^x \cdot 1 = e^x$$

```python
# Visualize the self-similarity property of e^x

x = np.linspace(-2, 3, 1000)
y = np.exp(x)
y_prime = np.exp(x)  # Same function!

# Create comparison visualization
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: Original function
ax1.plot(x, y, 'b-', linewidth=3, label='f(x) = e^x')
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')
ax1.set_title('Original Function: f(x) = e^x')
ax1.grid(True, alpha=0.3)
ax1.legend()
ax1.set_ylim(0, 20)

# Plot 2: Derivative function
ax2.plot(x, y_prime, 'r-', linewidth=3, label="f'(x) = e^x")
ax2.set_xlabel('x')
ax2.set_ylabel("f'(x)")
ax2.set_title("Derivative: f'(x) = e^x")
ax2.grid(True, alpha=0.3)
ax2.legend()
ax2.set_ylim(0, 20)

# Plot 3: Overlay both functions
ax3.plot(x, y, 'b-', linewidth=3, label='f(x) = e^x', alpha=0.7)
ax3.plot(x, y_prime, 'r--', linewidth=3, label="f'(x) = e^x", alpha=0.7)
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_title("Self-Similarity: f(x) and f'(x) are identical!")
ax3.grid(True, alpha=0.3)
ax3.legend()
ax3.set_ylim(0, 20)

# Plot 4: Tangent lines showing slope = value property
x_points = [-1, 0, 1, 2]
ax4.plot(x, y, 'b-', linewidth=3, label='f(x) = e^x')

for x_point in x_points:
    y_point = np.exp(x_point)
    slope = np.exp(x_point)  # slope equals the y-value!
    
    # Draw tangent line
    x_tangent = np.linspace(x_point - 0.5, x_point + 0.5, 100)
    y_tangent = slope * (x_tangent - x_point) + y_point
    
    ax4.plot(x_tangent, y_tangent, 'r--', alpha=0.8, linewidth=2)
    ax4.plot(x_point, y_point, 'ro', markersize=8)
    ax4.annotate(f'({x_point}, {y_point:.2f})\nslope = {slope:.2f}', 
                xy=(x_point, y_point), xytext=(x_point + 0.3, y_point + 1),
                fontsize=10, ha='center')

ax4.set_xlabel('x')
ax4.set_ylabel('f(x)')
ax4.set_title('Tangent Lines: Slope = Function Value')
ax4.grid(True, alpha=0.3)
ax4.legend()
ax4.set_xlim(-2, 3)
ax4.set_ylim(0, 20)

plt.tight_layout()
plt.show()

# Numerical verification
print("Verification: Self-Similarity Property of e^x")
print("="*50)
print("At any point x, the function value equals its derivative:")
print()

for x_val in [-1, 0, 0.5, 1, 1.5, 2]:
    function_value = np.exp(x_val)
    numerical_deriv = numerical_derivative(np.exp, x_val)
    error = abs(function_value - numerical_deriv)
    
    print(f"x = {x_val:4.1f} | f(x) = e^x = {function_value:8.4f} | f'(x) = {numerical_deriv:8.4f} | Error: {error:.2e}")

print("\nKey Insights:")
print("• The exponential function is its own derivative")
print("• At every point, the slope of the tangent line equals the y-coordinate")
print("• This makes e^x the 'natural' exponential function")
print("• Growth rate is always proportional to current size")
```

## Special Case 3: Trigonometric Functions - The Cyclic Pattern

Trigonometric functions exhibit a beautiful **cyclic pattern** in their derivatives. This pattern is fundamental to understanding oscillations, waves, and periodic phenomena in physics and engineering.

### The Cyclic Derivative Pattern

$$\frac{d}{dx}[\sin(x)] = \cos(x)$$
$$\frac{d}{dx}[\cos(x)] = -\sin(x)$$
$$\frac{d}{dx}[-\sin(x)] = -\cos(x)$$
$$\frac{d}{dx}[-\cos(x)] = \sin(x)$$

### The Complete Cycle
Starting with $\sin(x)$ and repeatedly taking derivatives:

$$\sin(x) \xrightarrow{\text{differentiate}} \cos(x) \xrightarrow{\text{differentiate}} -\sin(x) \xrightarrow{\text{differentiate}} -\cos(x) \xrightarrow{\text{differentiate}} \sin(x)$$

**After 4 derivatives, we return to the original function!**

### Key Insights

1. **Phase Shift**: The derivative of sine is cosine, which is sine shifted by 90° (π/2 radians)
2. **Sign Changes**: Notice the negative signs appearing - derivatives can flip the sign
3. **Periodic Behavior**: Every 4th derivative returns to the original function
4. **Physical Meaning**: 
   - Position → Velocity → Acceleration → Jerk → Position (in oscillatory motion)
   - $\sin(x)$ represents position, $\cos(x)$ represents velocity

### Why This Pattern Matters

- **Wave Motion**: Describes how waves propagate and oscillate
- **Simple Harmonic Motion**: Springs, pendulums, vibrations
- **AC Circuits**: Electrical engineering applications
- **Signal Processing**: Fourier analysis and frequency domain

```python
# Visualize the cyclic pattern of trigonometric derivatives

x = np.linspace(0, 4*np.pi, 1000)

# The four functions in the cycle
sin_x = np.sin(x)
cos_x = np.cos(x)
neg_sin_x = -np.sin(x)
neg_cos_x = -np.cos(x)

# Create comprehensive visualization
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Plot the cyclic pattern
ax1.plot(x, sin_x, 'b-', linewidth=3, label='sin(x)')
ax1.plot(x, cos_x, 'r--', linewidth=3, label="cos(x) = d/dx[sin(x)]")
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('Step 1: sin(x) → cos(x)')
ax1.grid(True, alpha=0.3)
ax1.legend()
ax1.set_xlim(0, 4*np.pi)
ax1.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi, 5*np.pi/2, 3*np.pi, 7*np.pi/2, 4*np.pi])
ax1.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π', '5π/2', '3π', '7π/2', '4π'])

ax2.plot(x, cos_x, 'r-', linewidth=3, label='cos(x)')
ax2.plot(x, neg_sin_x, 'g--', linewidth=3, label="-sin(x) = d/dx[cos(x)]")
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('Step 2: cos(x) → -sin(x)')
ax2.grid(True, alpha=0.3)
ax2.legend()
ax2.set_xlim(0, 4*np.pi)
ax2.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi, 5*np.pi/2, 3*np.pi, 7*np.pi/2, 4*np.pi])
ax2.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π', '5π/2', '3π', '7π/2', '4π'])

ax3.plot(x, neg_sin_x, 'g-', linewidth=3, label='-sin(x)')
ax3.plot(x, neg_cos_x, 'm--', linewidth=3, label="-cos(x) = d/dx[-sin(x)]")
ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_title('Step 3: -sin(x) → -cos(x)')
ax3.grid(True, alpha=0.3)
ax3.legend()
ax3.set_xlim(0, 4*np.pi)
ax3.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi, 5*np.pi/2, 3*np.pi, 7*np.pi/2, 4*np.pi])
ax3.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π', '5π/2', '3π', '7π/2', '4π'])

ax4.plot(x, neg_cos_x, 'm-', linewidth=3, label='-cos(x)')
ax4.plot(x, sin_x, 'b--', linewidth=3, label="sin(x) = d/dx[-cos(x)]")
ax4.set_xlabel('x')
ax4.set_ylabel('y')
ax4.set_title('Step 4: -cos(x) → sin(x) (Full Cycle!)')
ax4.grid(True, alpha=0.3)
ax4.legend()
ax4.set_xlim(0, 4*np.pi)
ax4.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi, 5*np.pi/2, 3*np.pi, 7*np.pi/2, 4*np.pi])
ax4.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π', '5π/2', '3π', '7π/2', '4π'])

plt.tight_layout()
plt.show()

# Show all four functions together
fig, ax = plt.subplots(1, 1, figsize=(14, 8))

ax.plot(x, sin_x, 'b-', linewidth=3, label='f₁(x) = sin(x)', alpha=0.8)
ax.plot(x, cos_x, 'r-', linewidth=3, label='f₂(x) = cos(x)', alpha=0.8)
ax.plot(x, neg_sin_x, 'g-', linewidth=3, label='f₃(x) = -sin(x)', alpha=0.8)
ax.plot(x, neg_cos_x, 'm-', linewidth=3, label='f₄(x) = -cos(x)', alpha=0.8)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('The Complete Trigonometric Derivative Cycle')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_xlim(0, 4*np.pi)
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi, 5*np.pi/2, 3*np.pi, 7*np.pi/2, 4*np.pi])
ax.set_xticklabels(['0', 'π/2', 'π', '3π/2', '2π', '5π/2', '3π', '7π/2', '4π'])

plt.show()

# Numerical verification of trigonometric derivatives
print("Verification: Trigonometric Derivative Cycle")
print("="*60)

def sin_func(x): return np.sin(x)
def cos_func(x): return np.cos(x)
def neg_sin_func(x): return -np.sin(x)
def neg_cos_func(x): return -np.cos(x)

test_points = [0, np.pi/4, np.pi/2, np.pi, 3*np.pi/2]
point_names = ['0', 'π/4', 'π/2', 'π', '3π/2']

print("Step 1: d/dx[sin(x)] = cos(x)")
for i, x_val in enumerate(test_points):
    analytical = np.cos(x_val)
    numerical = numerical_derivative(sin_func, x_val)
    error = abs(analytical - numerical)
    print(f"x = {point_names[i]:>5} | cos(x) = {analytical:8.4f} | Numerical = {numerical:8.4f} | Error: {error:.2e}")

print("\nStep 2: d/dx[cos(x)] = -sin(x)")
for i, x_val in enumerate(test_points):
    analytical = -np.sin(x_val)
    numerical = numerical_derivative(cos_func, x_val)
    error = abs(analytical - numerical)
    print(f"x = {point_names[i]:>5} | -sin(x) = {analytical:8.4f} | Numerical = {numerical:8.4f} | Error: {error:.2e}")

print("\nKey Properties:")
print("• Trigonometric functions have a 4-step derivative cycle")
print("• Each derivative is a 90° phase shift of the previous function")
print("• This pattern repeats infinitely: f''''(x) = f(x)")
print("• Physical interpretation: position → velocity → acceleration → jerk → position")
```

## Summary: Key Insights from Special Cases

In this notebook, we explored three fundamental special cases that reveal the beauty and power of differentiation:

### 🔄 **The Reciprocal Function** $f(x) = \frac{1}{x}$
- **Derivative**: $f'(x) = -\frac{1}{x^2}$
- **Key Insight**: Always negative (decreasing function)
- **Physical Meaning**: Inverse relationships (like pressure vs. volume)
- **Special Property**: Undefined at x = 0, creating a discontinuity

### 🚀 **The Exponential Function** $f(x) = e^x$
- **Derivative**: $f'(x) = e^x$ (self-similarity!)
- **Key Insight**: Growth rate equals current value
- **Physical Meaning**: Exponential growth/decay processes
- **Special Property**: Only function that is its own derivative

### 🌊 **Trigonometric Functions** $\sin(x)$ and $\cos(x)$
- **Cyclic Pattern**: $\sin(x) \to \cos(x) \to -\sin(x) \to -\cos(x) \to \sin(x)$
- **Key Insight**: 90° phase shifts with each differentiation
- **Physical Meaning**: Oscillatory motion, waves, vibrations
- **Special Property**: 4th derivative returns to original function

### 💡 **Universal Lessons**

1. **Pattern Recognition**: Each function family has characteristic derivative behaviors
2. **Physical Intuition**: Derivatives represent rates of change in real-world phenomena
3. **Mathematical Elegance**: Special functions reveal deep mathematical structures
4. **Practical Applications**: These patterns appear everywhere in science and engineering

### 🎯 **Why These Matter**

- **Engineering**: Oscillations, feedback systems, signal processing
- **Physics**: Motion, waves, energy transformations  
- **Economics**: Growth models, optimization problems
- **Biology**: Population dynamics, reaction kinetics
- **Data Science**: Activation functions, optimization algorithms

### 🔗 **Connection to Previous Concepts**

These special cases demonstrate the power of the **formal definition of derivatives** we learned earlier. Each follows from the fundamental limit:

$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

But their unique properties make them the "building blocks" for understanding more complex functions through combination rules (which we'll explore next!).
